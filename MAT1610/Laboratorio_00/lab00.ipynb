```python
# Importar librerías de la sesión
import numpy as np               # Para cálculos numéricos y vectores
import sympy as sp               # Para cálculo simbólico (álgebra exacta)
import matplotlib.pyplot as plt  # Para graficar
from ipywidgets import interact, FloatSlider # Para interactividad

# Configuración visual
sp.init_printing()
%matplotlib inline
try: plt.style.use('seaborn-v0_8-whitegrid')
except: plt.style.use('seaborn-whitegrid')
plt.rcParams['figure.figsize'] = (7, 4)
```

# **1. NumPy: Vectores y Arreglos Numéricos**

`NumPy` (Numerical Python) es la base de la computación científica. Nos permite crear **arreglos** (vectores y matrices) que son mucho más rápidos y eficientes que las listas estándar de Python. Su mayor ventaja en cálculo es que permite aplicar operaciones matemáticas a miles de números al mismo tiempo (vectorización).


### Ejemplo 1.1: Creación de arreglos y operaciones básicas
A diferencia de las listas, si multiplicas un arreglo de NumPy por 2, multiplicará cada elemento matemáticamente, no duplicará la lista.


```python
# Ejemplo 1.1: Operaciones vectorizadas
lista_normal = [1, 2, 3]
vector_numpy = np.array([1, 2, 3])

print("Lista normal multiplicada por 2: ", lista_normal * 2)
print("Vector NumPy multiplicado por 2: ", vector_numpy * 2)

# Operaciones matemáticas directas
print("Vector original al cuadrado:     ", vector_numpy ** 2)
```

### Ejemplo 1.2: Generación de dominios con `linspace`
Para graficar funciones continuas, necesitamos generar muchos puntos (el dominio $x$). `np.linspace(inicio, fin, cantidad)` crea un vector con números equiespaciados.


```python
# Ejemplo 1.2: np.linspace
# Generamos 5 puntos entre el 0 y el 10
puntos_pocos = np.linspace(0, 10, 5)
print("5 puntos equiespaciados:  ", puntos_pocos)

# En la práctica, usaremos unos 100 o 200 puntos para gráficas suaves
x_dominio = np.linspace(-5, 5, 100)
print(f"\nSe ha generado un dominio con {len(x_dominio)} puntos.")
print(f"Primeros 3 valores: {x_dominio[:3]}")
```

### Ejemplo 1.3: Funciones Matemáticas Universales
NumPy trae sus propias funciones trigonométricas y exponenciales (`np.sin()`, `np.cos()`, `np.exp()`) que operan sobre vectores enteros instantáneamente.


```python
# Ejemplo 1.3: Funciones de NumPy
x_angulos = np.array([0, np.pi/2, np.pi]) # [0, 90°, 180°]

seno_val = np.sin(x_angulos)
print("El seno de 0, pi/2 y pi es: ", np.round(seno_val, 4)) # Redondeamos a 4 decimales

# Exponencial
x_exp = np.array([0, 1, 2])
exp_val = np.exp(x_exp)
print("e^x evaluado en 0, 1 y 2 es: ", exp_val)
```

# **2. SymPy: Cálculo y Álgebra Simbólica**

Mientras NumPy trabaja con números aproximados (decimales), `SymPy` trabaja con **símbolos matemáticos exactos**, como si estuvieras resolviendo a mano en tu cuaderno. En SymPy, las fracciones se mantienen como fracciones y $\pi$ se mantiene como $\pi$.


### Ejemplo 2.1: Definiendo símbolos y expresiones
Para usar SymPy, primero debemos indicarle a Python qué letras serán variables abstractas matemáticas.


```python
# Ejemplo 2.1: Variables simbólicas
x, y = sp.symbols('x y')

# Creamos una expresión algebraica
expresion = (x + y)**2

print("Expresión creada:")
display(expresion) # display() formatea la salida como LaTeX

print("\nExpresión expandida algebraicamente:")
display(sp.expand(expresion))
```

### Ejemplo 2.2: Sustitución y Evaluación
Podemos sustituir valores dentro de una expresión simbólica usando el método `.subs()`.


```python
# Ejemplo 2.2: Evaluando expresiones
x = sp.Symbol('x')
f = x**2 - 5*x + 6

print("Función f(x):")
display(f)

# Evaluar en x = 3
valor_en_3 = f.subs(x, 3)
print(f"\nf(3) = {valor_en_3}")

# Evaluar simbólicamente: sustituir x por 'y + 1'
f_sustituida = f.subs(x, sp.Symbol('y') + 1)
print("\nf(y + 1) sin simplificar:")
display(f_sustituida)
```

### Ejemplo 2.3: Resolver Ecuaciones (Encontrar raíces)
El comando `sp.solve()` es nuestra herramienta principal para encontrar dónde una función se hace cero (raíces) o dónde dos curvas se intersecan.


```python
# Ejemplo 2.3: Resolviendo ecuaciones
x = sp.Symbol('x')
ecuacion_cuadratica = x**2 - 5*x + 6

# sp.solve asume automáticamente que la expresión está igualada a 0
raices = sp.solve(ecuacion_cuadratica, x)

print("Las raíces (ceros) de x^2 - 5x + 6 = 0 son:")
display(raices)

# Para igualar dos curvas, por ejemplo x^2 = 4, usamos sp.Eq()
interseccion = sp.solve(sp.Eq(x**2, 4), x)
print("\nLas intersecciones de x^2 y 4 son:")
display(interseccion)
```

# **3. Matplotlib: Gráficas y Visualización**

`Matplotlib` es la librería para graficar. El flujo de trabajo estándar en este curso será:
1. Crear un dominio numérico con `NumPy` ($x$).
2. Calcular la salida numérica ($y$).
3. Usar `plt.plot(x, y)` para dibujar.
4. Agregar títulos, etiquetas y leyendas.


### Ejemplo 3.1: Gráfica Básica
Trazaremos la función polinomial $y = x^2$.


```python
# Ejemplo 3.1: Trazado 2D básico
x_vals = np.linspace(-5, 5, 100) # 1. Dominio
y_vals = x_vals**2               # 2. Rango

plt.plot(x_vals, y_vals) # 3. Dibujar
plt.title("Mi primera gráfica: $y = x^2$")          # 4. Formato
plt.xlabel("Eje X")
plt.ylabel("Eje Y")
plt.show() # Mostrar el gráfico limpio
```

### Ejemplo 3.2: Múltiples curvas y Leyendas
A menudo necesitaremos comparar dos funciones (como una función y su derivada, o una función y su inversa).


```python
# Ejemplo 3.2: Comparando curvas
x_vals = np.linspace(0, 4*np.pi, 200)

y_seno = np.sin(x_vals)
y_coseno = np.cos(x_vals)

plt.plot(x_vals, y_seno, label='$\sin(x)$', color='C00')
plt.plot(x_vals, y_coseno, label='$\cos(x)$', color='C01', linestyle='--')

plt.title("Comparación de Ondas Trigonométricas")
plt.legend() # Muestra las etiquetas (label) en un cuadro
plt.axhline(0, color='black', linewidth=1) # Dibuja la línea del eje X = 0
plt.show()
```

### Ejemplo 3.3: Puntos de Interés (Scatter)
Para resaltar raíces, máximos, mínimos o puntos de tangencia, usamos `plt.scatter(x, y)` para dibujar puntos específicos sobre la curva continua.


```python
# Ejemplo 3.3: Resaltando puntos específicos
x_vals = np.linspace(-3, 3, 100)
y_vals = x_vals**3 - 3*x_vals

plt.plot(x_vals, y_vals, color='gray')

# Resaltamos los picos (Máximo y Mínimo local)
# Coordenadas conocidas: (-1, 2) y (1, -2)
puntos_x = [-1, 1]
puntos_y = [2, -2]

plt.scatter(puntos_x, puntos_y, color='red', s=80, label='Puntos Críticos', zorder=5)

plt.title("Gráfico con Puntos Destacados")
plt.legend()
plt.show()
```

# **4. ipywidgets: Gráficas Interactivas**

Para entender parámetros dinámicos, el módulo `ipywidgets` nos proporciona el decorador `@interact`. Esto permite crear controles deslizantes (sliders) para observar visualmente cómo cambia una función al modificar sus constantes.


### Ejemplo 4.1: Interactividad de Texto Simple
La idea central es crear una función en Python (usando `def`) que reciba un parámetro, y colocar `@interact` justo encima.


```python
# Ejemplo 4.1: Interact básico
@interact(multiplicador=FloatSlider(value=2.0, min=0.5, max=5.0, step=0.5))
def mostrar_multiplicacion(multiplicador):
    print(f"El valor base es 10.")
    print(f"10 x {multiplicador} = {10 * multiplicador}")
```

### Ejemplo 4.2: Animación de la Frecuencia
Veamos cómo usar la interactividad para modificar la frecuencia de una onda seno: $y = \sin(k \cdot x)$.


```python
# Ejemplo 4.2: Gráfico interactivo (1 variable)
x_vals = np.linspace(0, 10, 300)

@interact(k=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description='Frecuencia k:'))
def graficar_onda(k):
    y_vals = np.sin(k * x_vals)
    
    plt.figure(figsize=(8, 4))
    plt.plot(x_vals, y_vals, color='purple', lw=2)
    plt.ylim(-1.5, 1.5)
    plt.title(f"Onda: $\sin({k:.1f} x)$")
    plt.axhline(0, color='black', alpha=0.3)
    plt.show()
```

### Ejemplo 4.3: Controles Múltiples (Parábola Dinámica)
Podemos añadir múltiples sliders para controlar varios aspectos de una ecuación simultáneamente. Modelemos el vértice de una parábola: $y = a(x - h)^2 + k$.


```python
# Ejemplo 4.3: Gráfico con múltiples controles
x_vals = np.linspace(-10, 10, 200)

@interact(
    a=FloatSlider(value=1.0, min=-3.0, max=3.0, step=0.5, description='Apertura (a):'),
    h=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='Desp. X (h):'),
    k_y=FloatSlider(value=0.0, min=-5.0, max=5.0, step=0.5, description='Desp. Y (k):')
)
def graficar_parabola(a, h, k_y):
    y_vals = a * (x_vals - h)**2 + k_y
    
    plt.figure(figsize=(8, 5))
    plt.plot(x_vals, y_vals, color='darkgreen', lw=2)
    
    # Marcamos el vértice dinámicamente
    plt.scatter([h], [k_y], color='red', s=80, label=f'Vértice ({h}, {k_y})', zorder=5)
    
    plt.xlim(-10, 10)
    plt.ylim(-10, 10)
    plt.axhline(0, color='black', lw=0.5)
    plt.axvline(0, color='black', lw=0.5)
    plt.title(f"$y = {a}(x - {h})^2 + {k_y}$")
    plt.legend()
    plt.show()
```
