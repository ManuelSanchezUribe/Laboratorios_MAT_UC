# Definici&oacute;n de Funci&oacute;n

Una funcion es una regla que asigna exactamente un valor de salidad a cada valor de entrada. Cuando una funcion $f$ tiene un valor de entrada $x$, el valor de salida se escribe como $f(x)$. Esto se lee como ``$f$ de $x$''

En Cálculo, el objeto fundamental son las funciones. En Python, podemos definirlas usando la palabra clave def.
Como ejemplo, consideremos la función cuadrática del examen de diagnóstico de Stewart: 
$$f(x)=x^2−2x+1$$
Esta función representa una parábola.
<!-- > **Ejemplo de funciones:**
> - $f(x) = 2x-3$, funcion lineal, resulta de multiplicar la entrada  $x$ por 2 y luego restar $3$. Regresa el valor $2x-3$
> - $g(x) = x^2 - 4$, funcion cuadratica.
> - $h(x) = \sin(x)$, funcion sinusoidal. -->

```python
# Definicion de funcion f(x) - parabola
def f(x): 
    return x**2-2*x+1
# Analogamente podemos definir funciones de tipo lambda
f = lambda x: x**2-2*x+1
```

```python
# Evaluacion puntual de la funcion
punto_x = float(input("Ingrese el valor de entrada x ="))   
f_de_x = f(punto_x) 
print("El valor de salida f(x) =", f_de_x) 
```

```python
# Crear un dominio de x desde -2 hasta 4 con 10 puntos
x_valores = np.linspace(-2, 4, 10)
y_valores = f(x_valores)

print("   x  |   f(x)")
print("-" * 15)
for x, y in zip(x_valores, y_valores):
    print(f"{x:5.2f} | {y:5.2f}")
```

```python
# grafica de las funcion
x = np.linspace(-3, 5, 100) # Dominio restringido por Stewart
# plt.figure(figsize=(8, 5))
plt.plot(x, f(x), label='$f(x) = x^{2}-2x+1$', color='red', linewidth=2)
plt.axhline(0, color='black',linewidth=1)
plt.axvline(0, color='black',linewidth=1)
plt.title('Representación Visual de una Función Par&aacute;bola')
plt.xlabel('x')
plt.ylabel('g(x)')
plt.grid(True, linestyle='--')
plt.legend()
plt.show()
```

## **2. Dominio y Gráficas de Funciones Raíz**
El dominio es el conjunto de todas las entradas posibles. Para la función $g(x) = \sqrt{16 - x^2}$, el dominio está restringido al intervalo $[-4, 4]$.


```python
def g(x):
    # Usamos np.sqrt para soporte de arreglos
    return np.sqrt(16 - x**2)

# Definimos el dominio basado en el análisis matemático
x = np.linspace(-4, 4, 100)
y = g(x)

plt.plot(x, y, label='$g(x) = \\sqrt{16 - x^2}$', color='purple')
plt.fill_between(x, y, alpha=0.2, color='purple') # Resalta el área bajo la curva
plt.title("Gráfica de una función con dominio restringido")
plt.grid(True)
plt.legend()
plt.show()
```

## **3. Funciones Definidas a Trozos (Piecewise)**
Estas funciones cambian su fórmula dependiendo del valor de entrada [6]. 
Ejemplo:
$$f(x) = \begin{cases} 1 - x & \text{si } x \le 1 \\ x^2 & \text{si } x > 1 \end{cases}$$


```python
def f_trozos(x):
    return np.piecewise(x, [x <= 1, x > 1], [lambda x: 1 - x, lambda x: x**2])

x_vals = np.linspace(-2, 3, 200)
plt.plot(x_vals, f_trozos(x_vals), lw=2, color='orange')
plt.axvline(1, color='gray', linestyle='--')
plt.title("Función definida por secciones")
plt.show()
```

## **4. Funciones Trigonométricas y el Círculo Unitario**
En Cálculo, las medidas se expresan siempre en **radianes** [7, 8]. 
Si un punto $P(x, y)$ está en el círculo unitario ($x^2 + y^2 = 1$), sus coordenadas están dadas por $(\cos \theta, \sin \theta)$ [9].


-------

```python
theta = np.linspace(0, 2*np.pi, 100)
plt.figure(figsize=(5,5))
plt.plot(np.cos(theta), np.sin(theta), color='black') # Círculo

# Dibujar un ángulo de pi/3
angulo = np.pi/3
plt.plot([0, np.cos(angulo)], [0, np.sin(angulo)], 'ro-') 
plt.text(0.6, 0.9, r'$P(\cos \theta, \sin \theta)$', color='red')

plt.axhline(0, color='gray'); plt.axvline(0, color='gray')
plt.gca().set_aspect('equal')
plt.title("Relación con el Círculo Unitario")
plt.show()
```

```python
# Crear la tabla de valores notables (Stewart, Apéndice D) [10]
angulos_nombres = ['0', 'pi/6', 'pi/4', 'pi/3', 'pi/2']
angulos_valores = [0, np.pi/6, np.pi/4, np.pi/3, np.pi/2]

data = {
    'Ángulo (x)': angulos_nombres,
    'sen(x)': [np.sin(a) for a in angulos_valores],
    'cos(x)': [np.cos(a) for a in angulos_valores],
    'tan(x)': [np.tan(a) if not np.isclose(a, np.pi/2) else "Indefinido" for a in angulos_valores]
}

df_trig = pd.DataFrame(data)
df_trig.set_index('Ángulo (x)', inplace=True)

# Formatear números para mejor lectura
df_trig
```

# Comportamiento de Funciones


## Funciones comunes: Lineas, exponenciales, logaritmica, sinuoidal

### Funciones Trigonométricas
En el Cálculo, por convención, siempre se utilizan medidas en radianes
. Una función trigonométrica como f(x)=sin(x) se entiende como el seno de un ángulo cuya medida en radianes es x
.
Relación con el Círculo Unitario: Si consideramos un círculo de radio r=1 centrado en el origen (círculo unitario), cualquier punto P(x,y) sobre su circunferencia puede representarse en términos de un ángulo θ en posición estándar
. Las coordenadas del punto son: 
x=cos(θ)
 
y=sin(θ)
 Debido a que el punto está en el círculo unitario, se cumple la identidad fundamental: sin 
2
 (θ)+cos 
2
 (θ)=1
.


```python
# Creación del círculo unitario para visualizar la relación
theta = np.linspace(0, 2*np.pi, 100)
x_puntos = np.cos(theta)
y_puntos = np.sin(theta)

fig, ax = plt.subplots(figsize=(6,6))
ax.plot(x_puntos, y_puntos, color='black', label='Círculo Unitario $x^2 + y^2 = 1$')

# Dibujar un ángulo de ejemplo (pi/3)
angulo = np.pi/3
px, py = np.cos(angulo), np.sin(angulo)
ax.plot([0, px], [0, py], 'ro-') # Radio
ax.text(px+0.05, py+0.05, f'P(cos θ, sen θ)\n({px:.2f}, {py:.2f})', color='red')

ax.axhline(0, color='gray', lw=1); ax.axvline(0, color='gray', lw=1)
ax.set_xlim(-1.5, 1.5); ax.set_ylim(-1.5, 1.5)
ax.set_aspect('equal')
ax.set_title("Relación entre el ángulo y las coordenadas")
ax.grid(True, linestyle=':')
plt.legend()
plt.show()
```

## **4.3 Animación Interactiva: Del Círculo al Seno**

Una de las formas más poderosas de entender las funciones trigonométricas es observar cómo se "desenrolla" el círculo unitario para formar las ondas senoidales. 

En esta animación, puedes mover el control deslizante para cambiar el ángulo $\theta$:
1. **A la izquierda:** Verás el círculo unitario con el punto $P(\cos \theta, \sin \theta)$.
2. **A la derecha:** Verás cómo la altura (coordenada $y$) de ese punto se mapea directamente en la gráfica de la función seno.


-------------

```python
def animar_seno(theta_final):
    # Crear figura con dos subplots
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    # --- Subplot 1: Círculo Unitario ---
    t = np.linspace(0, 2*np.pi, 100)
    ax1.plot(np.cos(t), np.sin(t), color='gray', alpha=0.3) # Círculo de fondo
    
    # Punto P y Radio
    px, py = np.cos(theta_final), np.sin(theta_final)
    ax1.plot([0, px], [0, py], 'ro-') 
    ax1.plot(px, py, 'ro', markersize=8)
    
    # Línea guía horizontal (proyección de y)
    ax1.axhline(py, xmin=0.5, xmax=1, color='red', linestyle=':', alpha=0.6)
    
    ax1.set_xlim(-1.2, 1.2); ax1.set_ylim(-1.2, 1.2)
    ax1.set_aspect('equal')
    ax1.set_title(f"Círculo Unitario\n$P = (\cos({theta_final:.2f}), \sin({theta_final:.2f}))$")
    ax1.grid(True, alpha=0.2)

    # --- Subplot 2: Función Seno ---
    x_curva = np.linspace(0, 2*np.pi, 200)
    ax2.plot(x_curva, np.sin(x_curva), color='blue', alpha=0.3) # Curva completa de fondo
    
    # Progreso de la curva hasta theta_final
    x_progreso = np.linspace(0, theta_final, 100)
    ax2.plot(x_progreso, np.sin(x_progreso), color='blue', lw=2)
    ax2.plot(theta_final, py, 'ro') # Punto actual en la curva
    
    ax2.set_xlim(0, 2*np.pi)
    ax2.set_ylim(-1.2, 1.2)
    ax2.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
    ax2.set_xticklabels(['0', '$\pi/2$', '$\pi$', '$3\pi/2$', '$2\pi$'])
    ax2.set_title("Función $f(x) = \sin(x)$")
    ax2.grid(True, alpha=0.2)
    
    plt.tight_layout()
    plt.show()

# Crear el control interactivo
interact(animar_seno, 
         theta_final=widgets.FloatSlider(min=0, max=2*np.pi, step=0.1, value=1.0, 
                                         description='Ángulo θ:'))

```

```python
def animar_seno_escala_pi(theta_final):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    # --- Círculo Unitario ---
    t = np.linspace(0, 2*np.pi, 100)
    ax1.plot(np.cos(t), np.sin(t), color='gray', alpha=0.3)
    px, py = np.cos(theta_final), np.sin(theta_final)
    ax1.plot([0, px], [0, py], 'ro-', lw=2) 
    
    # Arco del ángulo
    if theta_final > 0:
        t_arco = np.linspace(0, theta_final, 50)
        ax1.plot(0.2 * np.cos(t_arco), 0.2 * np.sin(t_arco), color='orange', lw=2)
        ax1.text(0.3 * np.cos(theta_final/2), 0.3 * np.sin(theta_final/2), 
                 r'$\theta$', color='darkorange', fontsize=14)

    ax1.set_xlim(-1.2, 1.2); ax1.set_ylim(-1.2, 1.2); ax1.set_aspect('equal')
    ax1.set_title(f"Ángulo actual: {theta_final/np.pi:.2f} $\pi$ rad")

    # --- Función Seno con Escala en PI ---
    x_curva = np.linspace(0, 2*np.pi, 200)
    ax2.plot(x_curva, np.sin(x_curva), color='blue', alpha=0.2)
    x_progreso = np.linspace(0, theta_final, 100)
    ax2.plot(x_progreso, np.sin(x_progreso), color='blue', lw=2)
    ax2.plot(theta_final, py, 'ro')
    
    # Aplicar escala en pi
    ax2.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
    ax2.set_xticklabels(['0', r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'])
    
    ax2.set_xlim(0, 2*np.pi); ax2.set_ylim(-1.2, 1.2); ax2.grid(True, alpha=0.3)
    ax2.set_title("Mapeo al eje de radianes")
    
    plt.tight_layout()
    plt.show()

interact(animar_seno_escala_pi, 
         theta_final=widgets.FloatSlider(min=0, max=2*np.pi, step=0.05*np.pi, value=np.pi/4, 
                                         description='Ángulo θ:'))
```

## **4.2 La Función Tangente y sus Asíntotas**

La función tangente se define analíticamente como la razón entre el seno y el coseno:
$$\tan(x) = \frac{\sin(x)}{\cos(x)}$$

A diferencia del seno y el coseno, la tangente tiene un dominio restringido porque no está definida cuando $\cos(x) = 0$. Esto ocurre en los múltiplos impares de $\pi/2$ (por ejemplo, $\pm \pi/2, \pm 3\pi/2$), donde la función presenta **asíntotas verticales** [1-3].

**Propiedades clave:**
*   **Periodo:** Su periodo es $\pi$, lo que significa que se repite cada $\pi$ unidades [1, 4].
*   **Rango:** El rango de la función tangente es $(-\infty, \infty)$ [1, 4].
*   **Comportamiento:** A medida que $x$ se acerca a $\pi/2$ por la izquierda, los v

```python
# Crear un dominio de x de -2pi a 2pi con alta resolución
x = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
y = np.tan(x)

# Los dispositivos de graficación a menudo intentan unir los puntos de 
# infinito positivo a negativo, creando líneas verticales erróneas [6, 7].
# Para evitar esto, "limpiamos" los valores que se disparan en las asíntotas.
y[np.abs(y) > 10] = np.nan 

plt.figure(figsize=(10, 6))
plt.plot(x, y, label=r'$y = \tan(x)$', color='teal', lw=2)

# Dibujar las asíntotas verticales teóricas (donde cos(x) = 0) [2]
asintotas = [-3*np.pi/2, -np.pi/2, np.pi/2, 3*np.pi/2]
for a in asintotas:
    plt.axvline(a, color='red', linestyle='--', alpha=0.5, label='Asíntota' if a == np.pi/2 else "")

# Configuración de ejes y etiquetas
plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.ylim(-7, 7) # Restringimos el rango visual para apreciar la curva
plt.xticks([-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi],
           [r'$-2\pi$', r'$-\frac{3\pi}{2}$', r'$-\pi$', r'$-\frac{\pi}{2}$', '0', r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'])

plt.title("Visualización de la Función Tangente y sus Asíntotas Verticales")
plt.xlabel("x (radianes)")
plt.ylabel("tan(x)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

## Composicion de Funciones

```python
def explorar_transformacion(a, h, k):
    x = np.linspace(-5, 5, 400)
    y_base = np.sin(x) # Usaremos el seno como base
    y_trans = a * np.sin(x - h) + k
    
    plt.figure(figsize=(8, 5))
    plt.plot(x, y_base, '--', label='Original: sen(x)', color='gray', alpha=0.5)
    plt.plot(x, y_trans, label=f'Transformada', color='blue', lw=2)
    plt.ylim(-5, 5); plt.axhline(0, color='black'); plt.grid(True)
    plt.legend(); plt.show()

interact(explorar_transformacion, 
         a=widgets.FloatSlider(min=-2, max=3, step=0.5, value=1),
         h=widgets.FloatSlider(min=-3, max=3, step=0.5, value=0),
         k=widgets.FloatSlider(min=-3, max=3, step=0.5, value=0))
```

```python
#Utilizaremos un ejemplo basado en el texto de Stewart para mostrar que el orden de la composición importa.

# Definición de funciones base
def f(x):
    return x**2

def g(x):
    return x - 3

# Definición de funciones compuestas
def f_comp_g(x):
    return f(g(x))  # f(x-3) = (x-3)^2

def g_comp_f(x):
    return g(f(x))  # g(x^2) = x^2 - 3

# Visualización
x_vals = np.linspace(-2, 8, 400)

plt.figure(figsize=(10, 6))
plt.plot(x_vals, f_comp_g(x_vals), label=r'$(f \circ g)(x) = (x-3)^2$', lw=2)
plt.plot(x_vals, g_comp_f(x_vals), label=r'$(g \circ f)(x) = x^2 - 3$', lw=2, linestyle='--')
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.title("La composición no es conmutativa")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

## Inversa de Funciones

```python
# Usaremos SymPy para manejar la inversa de forma simbólica
x = sp.symbols('x')
f_expr = x**3 + 2

# Resolvemos y = x^3 + 2 para x -> f_inv = (x-2)^(1/3)
f_inv_expr = sp.root(x - 2, 3)

# Convertimos a funciones numéricas para graficar
f_num = sp.lambdify(x, f_expr, 'numpy')
f_inv_num = sp.lambdify(x, f_inv_expr, 'numpy')

# Rango de valores (evitamos problemas con raíces de negativos en algunos motores)
x_f = np.linspace(-2, 2, 200)
x_inv = np.linspace(-6, 10, 400)

plt.figure(figsize=(8, 8))
plt.plot(x_f, f_num(x_f), label=r'$f(x) = x^3 + 2$', color='blue', lw=2)
plt.plot(x_inv, f_inv_num(x_inv), label=r'$f^{-1}(x) = \sqrt[12]{x-2}$', color='red', lw=2)
plt.plot([-10, 10], [-10, 10], color='gray', linestyle=':', label='$y = x$') # Recta de simetría

plt.xlim(-8, 8); plt.ylim(-8, 8)
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.gca().set_aspect('equal')
plt.title("Simetría entre una función y su inversa")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
