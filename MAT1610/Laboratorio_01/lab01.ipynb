# **1. Definición y Representación Algebraica**

Una **función** $f$ es una regla que asigna a cada elemento $x$ de un conjunto $D$ exactamente un elemento, llamado $f(x)$, de un conjunto $E$. El número $f(x)$ es el **valor de $f$ en $x$** y se lee simplemente como "$f$ de $x$".

En Cálculo, las funciones son los objetos fundamentales de estudio. En Python, podemos definir estas "reglas" usando la palabra clave `def` o también usando funciones `lambda`. Como ejemplo, consideremos la siguiente función cuadrática: 

$$f(x) = x^2 - 2x + 1$$

Esta representación algebraica corresponde visualmente a una parábola.


```python
# 1. Definicion de la funcion
# Definición estándar
def f(x): 
    return x**2 - 2*x + 1

# Definición con funciones lambda
g = lambda x: x**2 - 2*x + 1
```

```python
# 2. Evaluacion puntual de la funcion
punto_x = float(input("Ingrese el valor de entrada x ="))   
print(f"El valor de salida f(x) = {f(punto_x):1.4f}, valor de salida de g(x) = {g(punto_x):1.4f}") 
```

```python
# Crear el dominio y generar la tabla de valores
x_vals = np.linspace(-2, 4, 6)
pd.DataFrame({'x': x_vals, 'f(x)': f(x_vals)}).round(2)
```

```python
# 3. Representación Visual (Gráfica)
x = np.linspace(-3, 5, 100)
plt.plot(x, f(x), label=r'$f(x) = x^2-2x+1$', color='red', lw=2)

# Dibujar ejes cartesianos
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)

# Etiquetas y formato
plt.title('Representación Visual: Parábola')
plt.xlabel(r'$x$'); plt.ylabel(r'$f(x)$')
plt.legend(); plt.show()
```

## **Dominio de una Función**

El **dominio** de una función es el conjunto de todas las entradas ($x$) posibles para las que la regla de la función tiene sentido y define un número real.

Por ejemplo, para la función:
$$f(x) = \sqrt{16 - x^2}$$

El dominio está restringido al intervalo **$[-4, 4]$**. Si eligiéramos un valor fuera de este rango (como $x=5$), el número dentro de la raíz sería negativo, lo cual no está definido en los números reales.



```python
x = np.linspace(-4, 4, 100)
f = lambda x: np.sqrt(16-x**2)
# Gráfica de la función
plt.plot(x, f(x), label=r'$f(x) = \sqrt{16 - x^2}$', color='purple', lw=2)

# Representación del dominio sobre el eje x
plt.hlines(0, -4, 4, color='purple', lw=10, alpha=0.5, label='Dominio: [-4, 4]')

# Formato de ejes
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.title("Representación Visual del Dominio"); plt.legend(); plt.show()
```

```python
# que pasa en python si escogemos x = 5?
print(f"El valor de salida f(x) = {f(5):1.4f}, (Not a Number)")
```

## **Funciones Definidas por Secciones (A trozos)**

Existen funciones que se definen mediante fórmulas distintas en diferentes tramos de su dominio. Para evaluar estas funciones, la regla es la siguiente: primero debemos observar el valor de la entrada $x$ para identificar en qué intervalo se encuentra y luego aplicar la fórmula correspondiente.

Consideremos el siguiente **ejemplo**:
$$f(x) = \begin{cases} 1 - x & \text{si } x \le 1 \\ x^2 & \text{si } x > 1 \end{cases}$$

Este tipo de funciones son fundamentales para modelar situaciones donde el comportamiento cambia de reglas al cruzar un umbral específico.



```python
x = np.linspace(-2, 3, 1000)

# Definición y graficación de la función por secciones
f_trozos = np.piecewise(x, [x <= 1, x > 1], [lambda x: 1 - x, lambda x: x**2])

plt.plot(x, f_trozos, lw=2, color='orange')
plt.axvline(1, color='gray', linestyle='--', alpha=0.5)
plt.xlim(-2,3)
plt.title("Gráfica de una Función Definida por Secciones"); plt.show()
```

# **2. Funciones Comunes**

## Funciones Lineales, Polinomiales y racionales 
Presentamos a continuacion una clasificaci&oacute;n de las funciones según su forma algebraica:

*   **Lineales:** Tienen la forma $f(x) = mx + b$. Crecen o decrecen a una razón constante.
*   **Polinomiales:** Funciones como $P(x) = a_n x^n + a_{n-1} x^{n-1}... +a_1 x + a_0$. Su dominio es siempre $(-\infty, \infty)$. El valor de $n$ determina su **grado**.
*   **Racionales:** Son cocientes de dos polinomios $f(x) = \frac{P(x)}{Q(x)}$. Su dominio está restringido a los valores donde el denominador $Q(x) \neq 0$.



```python
def rac(x): return (x**2 + 1) / (x**2 - 1)
x = np.linspace(-4, 4, 500)
y = rac(x)
y[np.abs(y) > 10] = np.nan 

plt.plot(x, y, color='green', lw=2, label=r'$f(x) = \frac{x^2+1}{x^2-1}$')

# ILUSTRACIÓN DEL DOMINIO: Segmentos con huecos en las asíntotas (-1 y 1)
plt.hlines(0, -4, -1.1, colors='green', lw=8, alpha=0.3)
plt.hlines(0, -0.9, 0.9, colors='green', lw=8, alpha=0.3)
plt.hlines(0, 1.1, 4, colors='green', lw=8, alpha=0.3, label=r'Dominio: $x \neq \pm 1$')

# Asíntotas verticales
plt.axvline(1, color='red', ls=':'); plt.axvline(-1, color='red', ls=':')

plt.ylim(-5, 5); plt.axhline(0, color='black', lw=0.5)
plt.xlim(-4, 4); plt.axhline(0, color='black', lw=0.5)
plt.title("Función Racional y visualización del Dominio"); plt.legend(); plt.show()
```

### **Función Exponencial**

Las funciones exponenciales tienen la forma $f(x) = a^x$, donde la base $a$ es una constante positiva. La función exponencial más importante en cálculo es la **función exponencial natural** $f(x) = e^x$, donde $e \approx 2.71828$.

*   **Dominio:** $(-\infty, \infty)$ (toda la recta real).
*   **Rango:** $(0, \infty)$ (siempre son positivas).
*   **Punto clave:** Todas pasan por el punto $(0, 1)$ porque $a^0 = 1$.


```python
x = np.linspace(-2, 3, 100)

# Gráfica de la función exponencial natural
plt.plot(x, np.exp(x), label='$f(x) = e^x$', lw=2)

# ILUSTRACIÓN DEL DOMINIO: Línea continua sobre el eje x
plt.hlines(0, -2, 3, colors='blue', lw=8, alpha=0.3, label='Dominio: $\mathbb{R}$')

plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.ylim(-1, 20); plt.axhline(0, color='black', lw=0.5)
plt.xlim(-2, 3); plt.axhline(0, color='black', lw=0.5)
plt.title("Función Exponencial Natural"); plt.legend(); plt.show()
```

### **Función Logaritmo**

Las funciones logarítmicas $f(x) = \log_a x$ son las **inversas** de las funciones exponenciales. El logaritmo natural, denotado como $\ln(x)$, es el que tiene como base el número $e$.

*   **Dominio:** $(0, \infty)$ (solo aceptan entradas positivas).
*   **Rango:** $(-\infty, \infty)$.
*   **Comportamiento:** Tienen una **asíntota vertical** en el eje $y$ ($x=0$) y pasan por el punto $(1, 0)$.

--------------------------------------------------------

```python
x = np.linspace(0.01, 5, 100) # Iniciamos cerca de 0 para evitar error

# Gráfica del logaritmo natural
plt.plot(x, np.log(x), label='$f(x) = \ln(x)$', color='orange', lw=2)

# ILUSTRACIÓN DEL DOMINIO: Solo la parte positiva del eje x
plt.hlines(0, 0, 5, colors='orange', lw=8, alpha=0.3, label='Dominio: $(0, \infty)$')

# Asíntota vertical en x = 0
plt.axvline(0, color='red', ls=':', label='Asíntota: $x=0$')

plt.axhline(0, color='black', lw=1)
plt.ylim(-5, 2); plt.axhline(0, color='black', lw=0.5)
plt.xlim(-0.25, 5); plt.axhline(0, color='black', lw=0.5)
plt.title("Función Logaritmo Natural"); plt.legend(); plt.show()
```

### **Funciones Trigonom&eacute;tricas**

Para un ángulo agudo $\theta$, las seis funciones trigonométricas se definen como las **razones entre las longitudes de los lados** de un triángulo rectángulo:

| Función | Definición | Función Recíproca | Definición |
| :---: | :---: | :---: | :---: |
| $\sin(\theta)$ | $\frac{\text{cateto opuesto}}{\text{hipotenusa}}$ | $\csc(\theta)$ | $\frac{\text{hipotenusa}}{\text{cateto opuesto}}$ |
| $\cos(\theta)$ | $\frac{\text{cateto adyacente}}{\text{hipotenusa}}$ | $\sec(\theta)$ | $\frac{\text{hipotenusa}}{\text{cateto adyacente}}$ |
| $\tan(\theta)$ | $\frac{\text{cateto puesto}}{\text{cateto adyacente}}$ | $\cot(\theta)$ | $\frac{\text{cateto adyacente}}{\text{cateto opuesto}}$ |



```python
# Dibujo del triángulo rectángulo 
plt.plot([0,2,2,0], [0,0,1,0], 'b', lw=2) # 
plt.text(1, -0.15, 'Cateto Adyacente', ha='center')
plt.text(2.1, 0.5, 'Cateto Opuesto', va='center', rotation=90)
plt.text(0.8, 0.55, 'Hipotenusa', rotation=26)
plt.text(0.3, 0.05, r'$\theta$', fontsize=15)
plt.axis('equal'); plt.axis('off');plt.show()
```

Esta definición es la base para el estudio de las funciones trigonométricas antes de extenderlas a ángulos generales mediante el círculo unitario.

Las funciones seno ($f(x) = \sin x$) y coseno ($f(x) = \cos x$) describen movimientos ondulatorios. Una de sus propiedades más importantes es la **periodicidad**; ambas tienen un periodo de $2\pi$, lo que significa que sus valores se repiten exactamente cada vuelta completa al círculo.

*   **Dominio:** $(-\infty, \infty)$.
*   **Rango:** $[-1, 1]$ (sus valores nunca exceden 1 ni son menores a -1).

```python
x = np.linspace(-2*np.pi, 2 * np.pi, 200)

plt.plot(x, np.sin(x), label='$\sin(x)$', lw=2)
plt.plot(x, np.cos(x), label='$\cos(x)$', lw=2, ls='--')

# Configuración de escala en pi
plt.xticks([-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi], 
           ['$-2\pi$', '$-3\pi/2$', '$-\pi$', '$-\pi/2$','0', '$\pi/2$', '$\pi$', '$3\pi/2$', '$2\pi$'])
plt.axhline(0, color='black', lw=1); plt.legend(); plt.show()
```

### **Animación Interactiva: Del Círculo al Seno**

En Cálculo, los ángulos se miden habitualmente en **radianes**. La relación fundamental es que una media vuelta al círculo ($180^\circ$) equivale exactamente a **$\pi$ radianes**.

En un **círculo unitario** (radio $r=1$), cualquier ángulo $\theta$ determina un punto $P$ cuyas coordenadas son precisamente el coseno y el seno de dicho ángulo: $P = (\cos \theta, \sin \theta)$.

Una de las formas más poderosas de entender las funciones trigonométricas es observar cómo se "desenrolla" el círculo unitario para formar las ondas senoidales. 

En esta animación, puedes mover el control deslizante para cambiar el ángulo $\theta$:
1. **A la izquierda:** Verás el círculo unitario con el punto $P(\cos \theta, \sin \theta)$.
2. **A la derecha:** Verás cómo la altura (coordenada $y$) de ese punto se mapea directamente en la gráfica de la función seno y con la del coseno.


-------------

```python
def animar_seno_escala_pi(theta_final):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))
    
    # --- Círculo Unitario ---
    t = np.linspace(0, 2*np.pi, 200)
    ax1.plot(np.cos(t), np.sin(t), color='gray', alpha=0.3)
    px, py = np.cos(theta_final), np.sin(theta_final)
    ax1.plot([0, px], [0, py], 'ro-', lw=2)  
    ax1.text(px, py, f' P({px:.2f}, {py:.2f})', fontsize=12)
    # Arco del ángulo
    if theta_final > 0:
        t_arco = np.linspace(0, theta_final, 50)
        ax1.plot(0.2 * np.cos(t_arco), 0.2 * np.sin(t_arco), color='C02', lw=2)
        ax1.text(0.3 * np.cos(theta_final/2), 0.3 * np.sin(theta_final/2), 
                 r'$\theta$', color='C02', fontsize=14)

    ax1.set_xlim(-1.2, 1.2); ax1.set_ylim(-1.2, 1.2); ax1.set_aspect('equal')
    ax1.set_title(f"Ángulo actual: {theta_final/np.pi:.2f} $\pi$ rad")

    # --- Función Seno con Escala en PI ---
    x_curva = np.linspace(0, 2*np.pi, 200)
    ax2.plot(x_curva, np.sin(x_curva), color='C00', alpha=0.2)
    x_progreso = np.linspace(0, theta_final, 100)
    ax2.plot(x_progreso, np.sin(x_progreso), color='C00', lw=2)
    ax2.plot(theta_final, py, 'ro')
    ax2.text(theta_final, py, f'({py:.2f})', fontsize=12)

    # --- Función Coseno con Escala en PI ---
    x_curva = np.linspace(0, 2*np.pi, 200)
    ax2.plot(x_curva, np.cos(x_curva), color='C01', alpha=0.2)
    x_progreso = np.linspace(0, theta_final, 100)
    ax2.plot(x_progreso, np.cos(x_progreso), color='C01', lw=2)
    ax2.plot(theta_final, px, 'ro')
    ax2.text(theta_final, px, f'({px:.2f})', fontsize=12)
    # Aplicar escala en pi
    ax2.set_xticks([0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi])
    ax2.set_xticklabels(['0', r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'])
    
    ax2.set_xlim(0, 2*np.pi); ax2.set_ylim(-1.2, 1.2); ax2.grid(True, alpha=0.3)
    ax2.set_title("Mapeo al eje de radianes")
    
    plt.tight_layout()
    plt.show()

interact(animar_seno_escala_pi, 
         theta_final=widgets.FloatSlider(min=0, max=2*np.pi, step=0.05*np.pi, value=np.pi/4, 
                                         description='Ángulo θ:'));
```

### **La Función Tangente y sus Asíntotas**

La función tangente se define analíticamente como la razón entre el seno y el coseno:
$$\tan(x) = \frac{\sin(x)}{\cos(x)}$$

A diferencia del seno y el coseno, la tangente tiene un dominio restringido porque no está definida cuando $\cos(x) = 0$. Esto ocurre en los múltiplos impares de $\pi/2$ (por ejemplo, $\pm \pi/2, \pm 3\pi/2$), donde la función presenta **asíntotas verticales** [1-3].

**Propiedades clave:**
*   **Periodo:** Su periodo es $\pi$, lo que significa que se repite cada $\pi$ unidades.
*   **Rango:** El rango de la función tangente es $(-\infty, \infty)$.
*   **Comportamiento:** A medida que $x$ se acerca a $\pi/2$ por la izquierda, los valores crecen indefinidamente. Si nos acecamos a $\pi/2$ por la derecha los valores decrecen indefinidamente. Esto se repite en los puntos $x=k \pi/2$ para $k$ numero entero.

```python
# Crear un dominio de x de -2pi a 2pi con alta resolución
x = np.linspace(-2 * np.pi, 2 * np.pi, 1000)
y = np.tan(x)

# Los dispositivos de graficación a menudo intentan unir los puntos de 
# infinito positivo a negativo, creando líneas verticales erróneas [6, 7].
# Para evitar esto, "limpiamos" los valores que se disparan en las asíntotas.
y[np.abs(y) > 10] = np.nan 

plt.figure(figsize=(10, 6))
plt.plot(x, y, label=r'$y = \tan(x)$', color='teal', lw=2)

# Dibujar las asíntotas verticales teóricas (donde cos(x) = 0) [2]
asintotas = [-3*np.pi/2, -np.pi/2, np.pi/2, 3*np.pi/2]
for a in asintotas:
    plt.axvline(a, color='red', linestyle='--', alpha=0.5, label='Asíntota' if a == np.pi/2 else "")

# Configuración de ejes y etiquetas
plt.axhline(0, color='black', lw=1)
plt.axvline(0, color='black', lw=1)
plt.ylim(-7, 7) # Restringimos el rango visual para apreciar la curva
plt.xticks([-2*np.pi, -3*np.pi/2, -np.pi, -np.pi/2, 0, np.pi/2, np.pi, 3*np.pi/2, 2*np.pi],
           [r'$-2\pi$', r'$-\frac{3\pi}{2}$', r'$-\pi$', r'$-\frac{\pi}{2}$', '0', r'$\frac{\pi}{2}$', r'$\pi$', r'$\frac{3\pi}{2}$', r'$2\pi$'])

plt.title("Visualización de la Función Tangente y sus Asíntotas Verticales")
plt.xlabel("x (radianes)")
plt.ylabel("tan(x)")
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend()
plt.show()
```

# Operaciones de Funciones

Al igual que los números, las funciones se pueden sumar, restar, multiplicar y dividir para crear nuevas funciones. Si tenemos dos funciones $f$ y $g$, el álgebra de funciones se define de la siguiente manera (para todos los valores de $x$ que pertenezcan a la intersección de sus dominios):

* **Suma:** $(f+g)(x) = f(x) + g(x)$
* **Resta:** $(f-g)(x) = f(x) - g(x)$
* **Multiplicación:** $(fg)(x) = f(x)g(x)$
* **División:** $\left(\frac{f}{g}\right)(x) = \frac{f(x)}{g(x)}$ (donde $g(x) \neq 0$)

Además, existe una operación muy importante llamada **Composición de funciones**, denotada por $f \circ g$, que consiste en evaluar una función dentro de la otra:
$$(f \circ g)(x) = f(g(x))$$

### Ejemplo Guiado: Álgebra y Composición usando SymPy

Podemos usar el motor simbólico de Python (`sympy`) para realizar estas operaciones algebraicamente. Definamos $f(x) = x^2$ y $g(x) = \sin(x)$.

```python
# Definir variables y funciones simbólicas
x = sp.symbols('x')
f_sym = x**2
g_sym = sp.sin(x)

# 1. Operaciones básicas
suma_sym = f_sym + g_sym
mult_sym = f_sym * g_sym

print(f"Suma (f+g)(x) = {suma_sym}")
print(f"Multiplicación (f*g)(x) = {mult_sym}")

# 2. Composición de funciones
# En SymPy, la composición f(g(x)) se hace sustituyendo 'x' en 'f' por la función 'g'
comp_fg_sym = f_sym.subs(x, g_sym)
comp_gf_sym = g_sym.subs(x, f_sym)

print(f"Composición (f o g)(x) = {comp_fg_sym}")
print(f"Composición (g o f)(x) = {comp_gf_sym}")
```

### Visualización Gráfica de la Suma de Funciones
Es muy ilustrativo ver qué significa gráficamente sumar dos funciones. El valor de $y$ de la nueva función en cada punto $x$ es literalmente la suma de las alturas de $f(x)$ y $g(x)$ en ese mismo punto.

```python
# Definir el dominio numérico
x_vals = np.linspace(-3, 3, 400)

# Evaluar funciones numéricamente
y_f = x_vals**2
y_g = np.sin(3*x_vals) # Multiplicamos por 3 para que tenga más oscilaciones
y_suma = y_f + y_g

# Graficar
plt.figure()
plt.plot(x_vals, y_f, '--', label='$f(x) = x^2$', color='blue', alpha=0.6)
plt.plot(x_vals, y_g, '--', label='$g(x) = \sin(3x)$', color='red', alpha=0.6)
plt.plot(x_vals, y_suma, '-', label='$(f+g)(x) = x^2 + \sin(3x)$', color='purple', linewidth=2)

plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.title('Suma Gráfica de Funciones', fontsize=14)
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

```python
#Utilizaremos un ejemplo basado en el texto de Stewart para mostrar que el orden de la composición importa.

# Definición de funciones base
def f(x):
    return x**2

def g(x):
    return x - 3

# Definición de funciones compuestas
def f_comp_g(x):
    return f(g(x))  # f(x-3) = (x-3)^2

def g_comp_f(x):
    return g(f(x))  # g(x^2) = x^2 - 3

# Visualización
x_vals = np.linspace(-2, 8, 400)

plt.figure()
plt.plot(x_vals, f_comp_g(x_vals), label=r'$(f \circ g)(x) = (x-3)^2$', lw=2)
plt.plot(x_vals, g_comp_f(x_vals), label=r'$(g \circ f)(x) = x^2 - 3$', lw=2, linestyle='--')
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.title("La composición no es conmutativa")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

```

## Inversa de Funciones

```python
# funcion y su inversa
f_num = lambda x: x**3 + 2
f_inv_num = lambda x: np.cbrt(x - 2)

# Rango de valores (evitamos problemas con raíces de negativos en algunos motores)
x_f = np.linspace(-3, 2, 200)
x_inv = np.linspace(-8, 10, 400)

plt.figure(figsize=(6,6))
plt.plot(x_f, f_num(x_f), label=r'$f(x) = x^3 + 2$', color='C00', lw=2)
plt.plot(x_inv, f_inv_num(x_inv), label=r'$f^{-1}(x) = (x-2)^{1/3}$', color='C01', lw=2)
plt.plot([-10, 10], [-10, 10], color='gray', linestyle=':', label='$y = x$') # Recta de simetría

plt.xlim(-8, 8); plt.ylim(-8, 8)
plt.axhline(0, color='black', lw=1); plt.axvline(0, color='black', lw=1)
plt.gca().set_aspect('equal')
plt.title("Simetría entre una función y su inversa")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
