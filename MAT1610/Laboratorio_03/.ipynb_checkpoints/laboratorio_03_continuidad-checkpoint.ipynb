```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Configuración de estilo para gráficos
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# Definimos la variable simbólica principal
x = sp.Symbol("x")
```

## 1. Definición Formal de Continuidad en un Punto
Una función $f(x)$ es continua en un número $a$ si y sólo si se cumplen las siguientes tres condiciones:
1. $f(a)$ está definida (es decir, $a$ pertenece al dominio de $f$).
2. $\lim_{x \to a} f(x)$ existe.
3. $\lim_{x \to a} f(x) = f(a)$.

Si alguna de estas condiciones falla, decimos que $f$ es discontinua en $a$.


### Ejemplo 1.1: Continuidad de un Polinomio (Perfil aerodinámico base)
Evaluemos la continuidad de $f(x) = x^3 - 2x^2 + x$ en $a = 1.5$.


```python
f1 = x**3 - 2*x**2 + x
a1 = 1.5
f_a = f1.subs(x, a1)
lim_f = sp.limit(f1, x, a1)
es_continua = (f_a == lim_f) and f_a.is_real
print(f"1. f({a1}) = {f_a}")
print(f"2. Limite cuando x -> {a1} = {lim_f}")
print(f"3. ¿Es continua? {es_continua}")
```

### Ejemplo 1.2: Remoción de Discontinuidad (Tolerancia de manufactura)
Sea $f(x) = \frac{x^2 - 1}{x - 1}$. Analicemos el punto $a = 1$.


```python
f2 = (x**2 - 1)/(x - 1)
a2 = 1
# Intentar evaluar directamente lanzará un error o dará NaN en sistemas numéricos.
lim_f2 = sp.limit(f2, x, a2)
print(f"El limite en x={a2} es: {lim_f2}. Como f(1) no está definida por división por cero, no es continua en su forma original.")
# Redefinimos para hacerla continua:
f2_continua = sp.Piecewise((f2, sp.Ne(x, 1)), (lim_f2, sp.Eq(x, 1)))
print("Función redefinida (continua):", f2_continua)
```

### Ejemplo 1.3: Análisis por la Izquierda y Derecha
Sea $f(x) = |x| / x$. Comprobemos la condición 2 en $a=0$.


```python
f3 = sp.Abs(x) / x
lim_izq = sp.limit(f3, x, 0, dir="-")
lim_der = sp.limit(f3, x, 0, dir="+")
print(f"Límite por la izquierda: {lim_izq}")
print(f"Límite por la derecha: {lim_der}")
print(f"¿Existe límite global? {\"Sí\" if lim_izq == lim_der else \"No (Discontinua)\"}")
```

## 2. Clasificación de Discontinuidades
En ingeniería, el tipo de discontinuidad nos habla del tipo de falla o evento:
* **Evitable (Agujero):** $\lim_{x \to a} f(x)$ existe, pero no es igual a $f(a)$ (o $f(a)$ no existe).
* **De Salto:** Los límites laterales $\lim_{x \to a^-} f(x)$ y $\lim_{x \to a^+} f(x)$ existen, pero son diferentes.
* **Infinita:** $\lim_{x \to a} f(x) = \pm \infty$ o alguno de sus límites laterales lo es.


### Ejemplo 2.1: Discontinuidad Evitable
Grafiquemos $f(x) = \frac{\sin(x)}{x}$ cerca de $x=0$.


```python
x_vals = np.linspace(-10, 10, 400)
# Usamos np.sinc, pero manualmente para evidenciar el problema
y_vals = np.where(x_vals == 0, np.nan, np.sin(x_vals)/x_vals)
plt.plot(x_vals, y_vals, label=r"$f(x)=\frac{\sin(x)}{x}$", color="blue")
plt.scatter([0], [1], facecolors="none", edgecolors="red", s=80, zorder=5, label="Discontinuidad Evitable (Límite=1)")
plt.legend()
plt.title("Discontinuidad Evitable en x=0")
plt.show()
```

### Ejemplo 2.2: Discontinuidad de Salto (Señal digital de un interruptor)
La función escalón de Heaviside $H(x)$.


```python
x_vals = np.linspace(-5, 5, 400)
y_vals = np.heaviside(x_vals, 1)
# Cortar la línea vertical para mostrar el salto
y_vals[np.abs(x_vals) < 0.05] = np.nan
plt.plot(x_vals, y_vals, color="purple", lw=2, label="Señal H(x)")
plt.scatter([0], [0], facecolors="none", edgecolors="purple", s=80)
plt.scatter([0], [1], color="purple", s=80)
plt.title("Discontinuidad de Salto en x=0")
plt.legend()
plt.show()
```

### Ejemplo 2.3: Discontinuidad Infinita (Resonancia magnética)
Analicemos la fuerza magnética $F(x) = \frac{1}{(x-2)^2}$ en $x=2$.


```python
x_vals = np.linspace(0, 4, 400)
# Enmascarar la asíntota
y_vals = 1 / (x_vals - 2)**2
y_vals[y_vals > 50] = np.nan
plt.plot(x_vals, y_vals, color="red")
plt.axvline(2, color="black", linestyle="--", label="Asíntota Vertical (Falla/Resonancia)")
plt.title("Discontinuidad Infinita en x=2")
plt.legend()
plt.show()
```

## 3. Continuidad en Intervalos y Funciones Compuestas
Una función es continua en un intervalo si lo es en cada número del intervalo. Además, si $g$ es continua en $a$ y $f$ es continua en $g(a)$, entonces la composición $(f \circ g)(x) = f(g(x))$ es continua en $a$.
Esto justifica la regla: $\lim_{x \to a} f(g(x)) = f(\lim_{x \to a} g(x))$.


### Ejemplo 3.1: Continuidad en intervalo cerrado
Dominio de $f(x) = \sqrt{4 - x^2}$. Continua en $[-2, 2]$.


```python
f4 = sp.sqrt(4 - x**2)
# Dominio
dominio = sp.calculus.util.continuous_domain(f4, x, sp.S.Reals)
print(f"El intervalo de continuidad es: {dominio}")
```

### Ejemplo 3.2: Función Compuesta
Composición de $f(x) = e^x$ y $g(x) = \sin(x)$ en $x = \pi$.


```python
f_base = sp.exp(x)
g_base = sp.sin(x)
compuesta = f_base.subs(x, g_base)
limite_compuesta = sp.limit(compuesta, x, sp.pi)
eval_compuesta = compuesta.subs(x, sp.pi)
print(f"Funcion compuesta f(g(x)): {compuesta}")
print(f"Límite = {limite_compuesta}, Evaluación = {eval_compuesta}")
```

### Ejemplo 3.3: Trazado en Dominio Continuo
Graficando la función compuesta $\ln(1 + \cos(x))$.


```python
x_vals = np.linspace(-2*np.pi, 2*np.pi, 400)
# Para x donde cos(x) = -1, la función no es continua.
y_vals = np.log(1 + np.cos(x_vals))
y_vals[np.isinf(y_vals)] = np.nan
plt.plot(x_vals, y_vals, color="green")
plt.title(r"Continuidad de la función compuesta $\ln(1 + \cos(x))$")
plt.show()
```
