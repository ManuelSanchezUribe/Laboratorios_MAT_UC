```python
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Configuración estética de los gráficos
plt.style.use("seaborn-v0_8-whitegrid")
plt.rcParams["figure.figsize"] = (8, 5)

# Declaración de la variable simbólica base
x = sp.Symbol("x")
```

## 1. Definición Formal de Continuidad en un Punto
Analíticamente, evaluar la continuidad de una trayectoria o señal requiere que verifiquemos de forma rigurosa las tres condiciones impuestas en el cálculo.
Una función $f(x)$ es continua en $x = a$ si y solo si:
$$\lim_{x \to a} f(x) = f(a)$$
Esto se desglosa operacionalmente en asegurar que $f(a)$ existe en $\mathbb{R}$, que los límites laterales coinciden en un valor finito común, y que dicho valor numérico coincide exactamente con la evaluación de la función.


### Ejemplo 1.1: Continuidad Simbólica de un Polinomio
Comprobemos las tres condiciones en $a = 2$ para la función continua $f(x) = x^3 - 4x + 1$.


```python
f1 = x**3 - 4*x + 1
a1 = 2
f_a1 = f1.subs(x, a1)
lim_f1 = sp.limit(f1, x, a1)
print(f"Condición 1 (Evaluación): f({a1}) = {f_a1}")
print(f"Condición 2 (Límite): lim = {lim_f1}")
print(f"Condición 3 (Coincidencia): ¿Es continua? {f_a1 == lim_f1}")
```

### Ejemplo 1.2: Remoción y Redefinición Algebraica
Evaluemos la función racional $f(x) = \frac{x^2 - 9}{x - 3}$ en $a = 3$, donde la evaluación directa falla.


```python
f2 = (x**2 - 9)/(x - 3)
a2 = 3
lim_f2 = sp.limit(f2, x, a2)
print(f"Límite analítico cuando x -> {a2}: {lim_f2}")
# Redefinimos la función de forma seccionada para forzar su continuidad
f2_redefinida = sp.Piecewise((f2, sp.Ne(x, a2)), (lim_f2, sp.Eq(x, a2)))
print("Función extendida continua:")
sp.pprint(f2_redefinida)
```

### Ejemplo 1.3: Verificación por Límites Laterales Exactos
Analicemos por qué falla el límite en el origen para la función de signo racional $f(x) = \frac{|x|}{x}$.


```python
f3 = sp.Abs(x) / x
lim_izq = sp.limit(f3, x, 0, dir="-")
lim_der = sp.limit(f3, x, 0, dir="+")
print(f"Límite lateral izquierdo: {lim_izq}")
print(f"Límite lateral derecho: {lim_der}")
print(f"¿Existe el límite ordinario? {lim_izq == lim_der}")
```

## 2. Clasificación de Discontinuidades
Si una o más de las condiciones de continuidad fallan, la naturaleza geométrica del quiebre se clasifica según el comportamiento de sus límites:
* **Evitable:** El límite existe y es finito, pero difiere de la función o esta no existe en ese punto.
* **De Salto:** Ambos límites laterales existen y son finitos, pero toman valores distintos.
* **Infinita (Asintótica):** Al menos uno de los límites laterales tiende a $\infty$ o $-\infty$.


### Ejemplo 2.1: Visualización de un "Agujero" (Evitable)
Grafiquemos $f(x) = \frac{\sin(x)}{x}$ mostrando de forma clara su discontinuidad evitable en el origen mediante un punto vacío.


```python
x_num = np.linspace(-8, 8, 300)
# Omitimos el cero exacto en el cálculo numérico para evitar errores de división
y_num = np.where(x_num == 0, np.nan, np.sin(x_num)/x_num)
plt.plot(x_num, y_num, color="#2980b9", lw=2, label=r"$f(x) = \frac{\sin(x)}{x}$")
# Dibujamos el marcador circular vacío en el punto de discontinuidad
plt.scatter([0], [1], facecolors="none", edgecolors="#e74c3c", s=100, zorder=5, label="Discontinuidad Evitable")
plt.title("Discontinuidad Evitable en el Origen")
plt.legend()
plt.show()
```

### Ejemplo 2.2: Trazado de Discontinuidades de Salto Finito
Grafiquemos de manera correcta la función escalón unitario, aislando el punto de quiebre para evitar líneas fantasmas oblicuas.


```python
x_step = np.linspace(-3, 3, 400)
y_step = np.heaviside(x_step, 1)
# Rompemos el arreglo para separar las líneas continuas
y_step[np.abs(x_step) < 0.02] = np.nan
plt.plot(x_step, y_step, color="#8e44ad", lw=2.5, label="Escalón de Heaviside")
plt.scatter([0], [0], facecolors="none", edgecolors="#8e44ad", s=80, zorder=4)
plt.scatter([0], [1], color="#8e44ad", s=80, zorder=4)
plt.title("Discontinuidad de Salto Finito")
plt.legend()
plt.show()
```

### Ejemplo 2.3: Ruptura Asintótica (Discontinuidad Infinita)
Visualicemos la discontinuidad asintótica de $f(x) = \frac{1}{x-1}$ aplicando máscaras de truncado en el eje vertical.


```python
x_asym = np.linspace(-1, 3, 500)
y_asym = 1 / (x_asym - 1)
# Truncamos los valores extremos para estabilizar la escala del gráfico
y_asym[np.abs(y_asym) > 40] = np.nan
plt.plot(x_asym, y_asym, color="#c0392b", lw=2)
plt.axvline(1, color="gray", linestyle="--", label="Asíntota Vertical (x=1)")
plt.ylim(-10, 10)
plt.title("Discontinuidad Infinita")
plt.legend()
plt.show()
```

## 3. Continuidad en Intervalos y Funciones Compuestas
Si dos funciones $g$ y $f$ son continuas en sus respectivos dominios, el Teorema de Composición asegura que si $g$ es continua en $a$ y $f$ es continua en $g(a)$, entonces:
$$\lim_{x \to a} f(g(x)) = f\left(\lim_{x \to a} g(x)\right) = f(g(a))$$
Esto permite calcular límites complejos evaluando "de adentro hacia afuera" secuencialmente.


### Ejemplo 3.1: Determinación Automática del Intervalo de Continuidad
Usemos las herramientas de conjuntos de SymPy para aislar el intervalo real cerrado donde la función irracional $f(x) = \sqrt{9 - x^2}$ es continua.


```python
f_irr = sp.sqrt(9 - x**2)
intervalo_cont = sp.calculus.util.continuous_domain(f_irr, x, sp.S.Reals)
print(f"La función es totalmente continua en el intervalo: {intervalo_cont}")
```

### Ejemplo 3.2: Composición Simbólica
Evaluemos la continuidad en $a = 0$ de la composición $f(g(x))$ donde $f(u) = \cos(u)$ y $g(x) = \frac{\pi}{x^2 + 1}$.


```python
g_expr = sp.pi / (x**2 + 1)
f_expr = sp.cos(sp.Symbol("u"))
comp_expr = f_expr.subs(sp.Symbol("u"), g_expr)
lim_comp = sp.limit(comp_expr, x, 0)
eval_comp = comp_expr.subs(x, 0)
print(f"Expresión compuesta f(g(x)): {comp_expr}")
print(f"Límite calculado: {lim_comp} | Evaluación: {eval_comp}")
print(f"¿Se preserva la continuidad en la composición? {lim_comp == eval_comp}")
```

### Ejemplo 3.3: Composición Racional Logarítmica
Analicemos gráficamente los puntos de quiebre de la función compuesta $f(x) = \ln(x^2 - 4)$ donde el argumento interior se vuelve menor o igual a cero.


```python
x_comp = np.linspace(-5, 5, 500)
# El dominio válido requiere |x| > 2
y_comp = np.log(x_comp**2 - 4)
y_comp[x_comp**2 - 4 <= 0] = np.nan
plt.plot(x_comp, y_comp, color="#16a085", lw=2, label=r"$f(x) = \ln(x^2 - 4)$")
plt.title("Ruptura de Dominio en Funciones Compuestas Logarítmicas")
plt.legend()
plt.show()
```
