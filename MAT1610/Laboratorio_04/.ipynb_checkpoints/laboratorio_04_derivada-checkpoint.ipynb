```python
# Importar librerías de la sesión
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# Configuración para visualización matemática
sp.init_printing()
%matplotlib inline

# Configuración de estilo
try: plt.style.use('seaborn-v0_8-whitegrid')
except: plt.style.use('seaborn-whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

# **1. Derivadas, Notación y Propiedades**

La derivada de una función $f(x)$, denotada como $f'(x)$ (Lagrange) o $\frac{df}{dx}$ (Leibniz), es el límite de la razón de cambio promedio. Su definición formal es:
$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

Para no calcular límites infinitamente, se han demostrado **reglas de derivación** (potencia, producto, cociente, regla de la cadena). Computacionalmente, `SymPy` implementa estas reglas exactas a través del comando `sp.diff()`.


### Ejemplo 1.1: Cálculo Analítico y Derivadas de Orden Superior
Calculemos la primera y segunda derivada de la función de un oscilador amortiguado: $y(t) = e^{-0.5t} \cos(3t)$.


```python
# Ejemplo 1.1: Derivadas de orden superior con SymPy
t = sp.Symbol('t')
y = sp.exp(-0.5*t) * sp.cos(3*t)

# Primera derivada (Velocidad)
dy_dt = sp.diff(y, t)

# Segunda derivada (Aceleración) - Se puede indicar el orden '2'
d2y_dt2 = sp.diff(y, t, 2)

print("Función original y(t):")
display(y)
print("\nPrimera Derivada y'(t):")
display(dy_dt)
print("\nSegunda Derivada y''(t):")
display(d2y_dt2)
```

# **2. Grafos de Derivadas**

La gráfica de $f'(x)$ nos da una "radiografía" del comportamiento de $f(x)$:
* Cuando $f'(x) > 0$, la gráfica de $f(x)$ **sube** (es creciente).
* Cuando $f'(x) < 0$, la gráfica de $f(x)$ **baja** (es decreciente).
* Cuando $f'(x) = 0$, la curva original tiene una **tangente horizontal** (un posible punto de retorno o valle).


### Ejemplo 2.1: Análisis Gráfico Apilado
Dada la posición $s(t) = t^3 - 6t^2 + 9t$, graficaremos la posición y la velocidad ($v(t) = s'(t)$) en paneles apilados para visualizar cómo los ceros de la velocidad dictan los picos/valles de la posición.


```python
# Ejemplo 2.1: Posición vs Velocidad
t = sp.Symbol('t')
s_expr = t**3 - 6*t**2 + 9*t
v_expr = sp.diff(s_expr, t)

# Convertir a funciones evaluables por NumPy
s_num = sp.lambdify(t, s_expr, 'numpy')
v_num = sp.lambdify(t, v_expr, 'numpy')

t_vals = np.linspace(-0.5, 4.5, 200)
s_vals = s_num(t_vals)
v_vals = v_num(t_vals)

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 7))

# Panel de Posición
ax1.plot(t_vals, s_vals, color='blue', lw=2, label='$s(t)$ (Posición)')
ax1.set_ylabel('Posición (m)')
ax1.legend()

# Panel de Velocidad
ax2.plot(t_vals, v_vals, color='red', lw=2, label='$v(t) = s\'(t)$ (Velocidad)')
ax2.axhline(0, color='black', lw=1, linestyle='--')
ax2.set_ylabel('Velocidad (m/s)')
ax2.set_xlabel('Tiempo $t$ (s)')
ax2.legend()

# Marcar correspondencias (Puntos donde v(t) = 0)
raices_v = sp.solve(v_expr, t)
for raiz in raices_v:
    r_val = float(raiz)
    ax1.axvline(r_val, color='green', linestyle=':', alpha=0.7)
    ax2.axvline(r_val, color='green', linestyle=':', alpha=0.7)
    ax2.scatter([r_val], [0], color='green', zorder=5)

plt.suptitle('Relación Cinemática: Ceros de la Derivada = Retornos de la Función')
plt.tight_layout()
plt.show()
```

# **3. Valores Extremos y Optimización**

El **Teorema de Fermat** establece que si una función continua tiene un máximo o mínimo local en $c$, y su derivada existe allí, entonces $f'(c) = 0$. Estos puntos se denominan **puntos críticos**.

En el diseño en ingeniería (minimizar costos de material, maximizar el volumen de un contenedor), la optimización consiste en encontrar estos puntos críticos y clasificarlos.


### Ejemplo 3.1: Encontrando y Clasificando Extremos
Encuentre los puntos críticos de $f(x) = \frac{1}{4}x^4 - x^3 - 2x^2 + 3$ y clasifíquelos usando la **Prueba de la Segunda Derivada** ($f''(c) > 0 \implies$ mínimo, $f''(c) < 0 \implies$ máximo).


```python
# Ejemplo 3.1: Optimización de Puntos Críticos
x = sp.Symbol('x')
f = (1/4)*x**4 - x**3 - 2*x**2 + 3

# 1. Primera y segunda derivada
df = sp.diff(f, x)
d2f = sp.diff(df, x)

# 2. Puntos críticos (df/dx = 0)
puntos_criticos = sp.solve(df, x)
print(f"Puntos críticos encontrados en x = {puntos_criticos}\n")

# 3. Clasificación
for pc in puntos_criticos:
    eval_segunda = d2f.subs(x, pc)
    valor_f = f.subs(x, pc)
    
    if eval_segunda > 0:
        tipo = "Mínimo local (Cóncava hacia arriba)"
    elif eval_segunda < 0:
        tipo = "Máximo local (Cóncava hacia abajo)"
    else:
        tipo = "Prueba no concluyente"
        
    print(f"En x = {pc}: f''({pc}) = {eval_segunda} --> {tipo}")
    print(f"   -> Valor de la función: y = {valor_f}\n")
```

# **4. El Teorema del Valor Medio (TVM)**

El **Teorema del Valor Medio** conecta el cambio promedio sobre un intervalo con el cambio instantáneo en un punto interior. Si $f$ es continua en $[a, b]$ y derivable en $(a, b)$, entonces existe al menos un número $c$ en $(a, b)$ tal que:
$$f'(c) = \frac{f(b) - f(a)}{b - a}$$

Geométricamente significa que **existe al menos un punto en la curva donde la recta tangente es paralela a la recta secante** que une los extremos del intervalo.


### Ejemplo 4.1: Paralelismo garantizado por el TVM
Verifiquemos el TVM para $f(x) = x^3 - x$ en el intervalo $[0, 2]$. Encontraremos el valor $c$ garantizado y lo graficaremos.


```python
# Ejemplo 4.1: Teorema del Valor Medio
x = sp.Symbol('x')
f = x**3 - x
a, b = 0, 2

# 1. Pendiente de la recta secante (Tasa promedio)
f_a, f_b = f.subs(x, a), f.subs(x, b)
m_secante = (f_b - f_a) / (b - a)

# 2. Resolver f'(c) = m_secante
df = sp.diff(f, x)
ecuacion_tvm = sp.Eq(df, m_secante)
soluciones_c = sp.solve(ecuacion_tvm, x)

# Filtrar las soluciones para que estén dentro de (a, b)
c_valores = [c for c in soluciones_c if a < float(c) < b]
c_val = float(c_valores[0])
print(f"La tasa promedio es {m_secante}. El punto garantizado por el TVM en (0, 2) es c = {c_val:.4f}")

# --- Visualización ---
x_vals = np.linspace(-0.5, 2.5, 100)
f_num = sp.lambdify(x, f, 'numpy')

plt.plot(x_vals, f_num(x_vals), label='$f(x) = x^3 - x$', color='teal')

# Recta Secante
secante = float(m_secante) * (x_vals - a) + float(f_a)
plt.plot(x_vals, secante, 'k--', label='Recta Secante promedio')
plt.scatter([a, b], [f_a, f_b], color='black', zorder=5)

# Recta Tangente en c
f_c = f_num(c_val)
tangente = float(m_secante) * (x_vals - c_val) + f_c
plt.plot(x_vals, tangente, 'r-', label=f'Tangente paralela en $c \approx {c_val:.2f}$')
plt.scatter([c_val], [f_c], color='red', zorder=5)

plt.title('Teorema del Valor Medio')
plt.legend()
plt.show()
```
