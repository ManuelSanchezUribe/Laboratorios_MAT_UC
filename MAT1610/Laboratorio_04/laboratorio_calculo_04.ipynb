# **1. Derivadas, Notación y Propiedades**

La derivada de una función $f(x)$ respecto a $x$, denotada clásicamente como $f'(x)$ (notación de Lagrange) o $\frac{df}{dx}$ (notación de Leibniz), representa la tasa de cambio instantánea o geométricamente, la pendiente de la recta tangente a la curva en un punto.

Formalmente se define mediante el límite:
$$f'(x) = \lim_{h \to 0} \frac{f(x+h) - f(x)}{h}$$

Python, a través de `SymPy`, aplica automáticamente las propiedades que derivan de este límite (regla del producto, cociente y de la cadena) usando la función `sp.diff()`.

### Ejemplo 1.1: Cálculo Analítico y Notación
Calculemos la primera y segunda derivada de una función que modela un oscilador amortiguado: $f(t) = e^{-0.5t} \cos(3t)$.

```python
# Ejemplo 1.1: Cálculo de derivadas con SymPy
t = sp.Symbol('t')
f = sp.exp(-0.5*t) * sp.cos(3*t)

# Primera derivada (Velocidad si f es posición)
df_dt = sp.diff(f, t)

# Segunda derivada (Aceleración)
# Podemos poner (f, t, 2) o derivar la derivada
d2f_dt2 = sp.diff(f, t, 2)

print("Función original f(t):")
display(f)
print("\nPrimera Derivada f'(t):")
display(df_dt)
print("\nSegunda Derivada f''(t):")
display(d2f_dt2)
```

# **2. Gráficos de Derivadas**

Existe una profunda conexión geométrica entre una función y su derivada:
* Si $f'(x) > 0$ en un intervalo, entonces $f(x)$ es **creciente** en ese intervalo.
* Si $f'(x) < 0$ en un intervalo, entonces $f(x)$ es **decreciente**.
* Si $f'(x) = 0$, la recta tangente es horizontal (potencial máximo o mínimo).

Visualizar $f(x)$ y $f'(x)$ alineadas verticalmente es una técnica poderosa para comprender el comportamiento cinemático o físico de un modelo.

### Ejemplo 2.1: Análisis Gráfico Apilado
Dada la función de posición $s(t) = t^3 - 6t^2 + 9t + 1$, trazaremos la posición $s(t)$ y la velocidad $v(t) = s'(t)$ para observar cómo los ceros de la velocidad se alinean con los puntos de retorno (valles y picos) de la posición.

```python
# Ejemplo 2.1: Posición vs Velocidad
t = sp.Symbol('t')
s_expr = t**3 - 6*t**2 + 9*t + 1
v_expr = sp.diff(s_expr, t)

# Convertimos expresiones simbólicas a funciones evaluables de NumPy
s_num = sp.lambdify(t, s_expr, 'numpy')
v_num = sp.lambdify(t, v_expr, 'numpy')

t_vals = np.linspace(0, 5, 200)
s_vals = s_num(t_vals)
v_vals = v_num(t_vals)

# Creamos un gráfico con dos sub-paneles (subplots)
fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True, figsize=(8, 7))

# Panel Posición
ax1.plot(t_vals, s_vals, color='blue', label='$s(t)$ (Posición)')
ax1.set_ylabel('Posición')
ax1.legend()
ax1.grid(True)

# Panel Velocidad
ax2.plot(t_vals, v_vals, color='red', label='$v(t) = s\'(t)$ (Velocidad)')
ax2.axhline(0, color='black', linewidth=1, linestyle='--')
ax2.set_xlabel('Tiempo $t$')
ax2.set_ylabel('Velocidad')
ax2.legend()
ax2.grid(True)

# Marcamos las líneas donde la velocidad es cero
raices_v = sp.solve(v_expr, t)
for raiz in raices_v:
    r_val = float(raiz)
    ax1.axvline(r_val, color='green', linestyle=':', alpha=0.6)
    ax2.axvline(r_val, color='green', linestyle=':', alpha=0.6)
    ax2.scatter([r_val], [0], color='green', zorder=5)

plt.suptitle('Relación Cinemática: Ceros de la Derivada = Retornos de Posición')
plt.show()
```

# **3. Valores Extremos y Puntos Críticos**

El **Teorema de Fermat** establece que si $f$ tiene un extremo local (máximo o mínimo) en $c$, y si $f'(c)$ existe, entonces $f'(c) = 0$. Estos puntos se llaman **puntos críticos**.

En ingeniería, esto es el pilar de la **Optimización**: minimizar el costo de materiales de una viga, maximizar la eficiencia de un motor, o encontrar el punto de mayor estrés térmico en un circuito.

### Ejemplo 3.1: Encontrando y Clasificando Extremos
Encuentre los puntos críticos de $f(x) = \frac{1}{4}x^4 - \frac{4}{3}x^3 + 2x^2$ y clasifíquelos usando la prueba de la segunda derivada:
* Si $f''(c) > 0$, la gráfica es cóncava hacia arriba $\rightarrow$ **Mínimo local**.
* Si $f''(c) < 0$, la gráfica es cóncava hacia abajo $\rightarrow$ **Máximo local**.

```python
# Ejemplo 3.1: Optimización y Puntos Críticos
x = sp.Symbol('x')
f = (1/4)*x**4 - (4/3)*x**3 + 2*x**2

# 1. Calculamos derivadas
df = sp.diff(f, x)
d2f = sp.diff(df, x)

# 2. Encontramos puntos críticos (df = 0)
puntos_criticos = sp.solve(df, x)
print(f"Puntos críticos encontrados en x = {puntos_criticos}\n")

# 3. Clasificación usando f''(x)
for pc in puntos_criticos:
    eval_segunda = d2f.subs(x, pc)
    valor_f = f.subs(x, pc)
    
    if eval_segunda > 0:
        tipo = "Mínimo local"
    elif eval_segunda < 0:
        tipo = "Máximo local"
    else:
        tipo = "Prueba no concluyente"
        
    print(f"En x = {pc}: f''({pc}) = {eval_segunda} --> {tipo} (Valor: y = {valor_f:.2f})")
```

# **4. El Teorema del Valor Medio (TVM)**

El **Teorema del Valor Medio** es uno de los teoremas más teóricos pero a la vez más prácticos del Cálculo. Afirma que si $f$ es continua en $[a, b]$ y derivable en $(a, b)$, entonces existe al menos un punto $c$ en $(a, b)$ tal que:

$$f'(c) = \frac{f(b) - f(a)}{b - a}$$

**Interpretación Física:** La tasa de cambio instantánea en algún instante $c$ (la tangente) debe ser exactamente igual a la tasa de cambio promedio sobre el intervalo entero (la secante). Si viajaste 100 km en 1 hora, en algún instante de ese viaje tu velocímetro marcó *exactamente* 100 km/h.

### Ejemplo 4.1: Demostración Analítica y Gráfica del TVM
Verifiquemos el TVM para la función $f(x) = x^3 - 2x$ en el intervalo $[-2, 2]$. Buscaremos el valor(es) de $c$ y graficaremos la paralela entre la tangente y la secante.

```python
# Ejemplo 4.1: Teorema del Valor Medio
x = sp.Symbol('x')
f = x**3 - 2*x
a, b = -2, 2

# 1. Calculamos la tasa de cambio promedio (Secante)
f_a = f.subs(x, a)
f_b = f.subs(x, b)
m_secante = (f_b - f_a) / (b - a)
print(f"Pendiente de la secante promedio: {m_secante}")

# 2. Derivamos y resolvemos f'(c) = m_secante
df = sp.diff(f, x)
ecuacion_tvm = sp.Eq(df, m_secante)
puntos_c = sp.solve(ecuacion_tvm, x)

print(f"Valores de c garantizados por el TVM: {puntos_c}")

# --- VISUALIZACIÓN ---
# Seleccionamos el primer punto c positivo
c_val = float(puntos_c[1]) 
f_c = float(f.subs(x, c_val))

x_vals = np.linspace(-2.5, 2.5, 200)
f_num = sp.lambdify(x, f, 'numpy')
y_vals = f_num(x_vals)

plt.plot(x_vals, y_vals, label='$f(x) = x^3 - 2x$')

# Recta Secante: y - f(a) = m_secante * (x - a)
secante_vals = float(m_secante)*(x_vals - a) + float(f_a)
plt.plot(x_vals, secante_vals, 'k--', label='Recta Secante')

# Recta Tangente en c: y - f(c) = f'(c) * (x - c)
# Ojo: f'(c) es por definición igual a m_secante
tangente_vals = float(m_secante)*(x_vals - c_val) + f_c
plt.plot(x_vals, tangente_vals, 'r-', label=f'Recta Tangente en $c \approx {c_val:.2f}$')

plt.scatter([a, b], [f_a, f_b], color='black', zorder=5)
plt.scatter([c_val], [f_c], color='red', zorder=5)

plt.title('Teorema del Valor Medio: Paralelismo Secante-Tangente')
plt.legend()
plt.show()
```
