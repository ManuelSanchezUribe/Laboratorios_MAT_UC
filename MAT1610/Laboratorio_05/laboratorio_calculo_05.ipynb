# 1. Medidas de cambio en un intervalo y en un punto
La **razón de cambio promedio** mide el comportamiento en un intervalo $[t_1, t_2]$: $\frac{f(t_2) - f(t_1)}{t_2 - t_1}$. La **razón de cambio instantánea** es el límite de este cociente cuando el intervalo tiende a cero, definiendo la derivada en un punto.

### Ejemplo: Velocidad promedio vs instantánea
Analizaremos un móvil cuya posición es $s(t) = t^3 - 2t^2$. Calcularemos la velocidad promedio en $[1, 2]$ y la velocidad instantánea en $t=1$.

# 2. Líneas secante y tangente
La pendiente de la **recta secante** representa la razón promedio. A medida que el punto $x_2$ se acerca a $x_1$, la secante converge a la **recta tangente**, que nos da la razón instantánea. Este es el concepto geométrico fundamental de la derivada.

# 3. Linealización
La **linealización** de una función $f(x)$ cerca de un punto $a$ es la aproximación por su recta tangente:
$$L(x) = f(a) + f'(a)(x-a)$$
Esto permite sustituir funciones complejas por polinomios de grado 1 para cálculos rápidos.

# 4. Razones de cambio y diferenciales
El diferencial $dy$ representa el cambio aproximado en $y$ ante un cambio pequeño $dx$:
$$dy = f'(x) dx$$
Es crucial para estimar errores en mediciones físicas: si una regla tiene una incertidumbre $\Delta x = dx$, el diferencial nos da la incertidumbre propagada en el resultado final.
