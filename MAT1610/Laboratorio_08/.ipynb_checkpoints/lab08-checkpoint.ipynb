```python
# Importar librerías de la sesión
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from scipy.integrate import cumtrapz
import warnings
warnings.filterwarnings('ignore')

# Configuración para visualización matemática
sp.init_printing()
%matplotlib inline

# Configuración de estilo
try: plt.style.use('seaborn-v0_8-whitegrid')
except: plt.style.use('seaborn-whitegrid')
plt.rcParams['figure.figsize'] = (8, 5)
```

# **1. Sumas de Riemann: Puntos Izquierdo, Derecho y Medio**

La base geométrica de la integral es el cálculo del área bajo una curva. Como ingenieros, no siempre tenemos funciones perfectas, por lo que aproximar áreas es vital. Dividimos el intervalo $[a, b]$ en $n$ subintervalos de ancho $\Delta x = \frac{b-a}{n}$. 

El área aproximada se calcula sumando el área de los rectángulos:
$$ A \approx \sum_{i=1}^{n} f(x_i^*) \Delta x $$

Donde $x_i^*$ es un punto de muestra en el subintervalo. Los tres métodos más comunes son:
1. **Punto Derecho:** $x_i^* = a + i\Delta x$
2. **Punto Izquierdo:** $x_i^* = a + (i-1)\Delta x$
3. **Punto Medio:** $x_i^* = a + (i - 0.5)\Delta x$

### Ejemplo 1.1: Aproximación por el Punto Derecho

```python
# Definimos nuestra función
f = lambda x: np.sin(x) + 1.5
a, b = 0, 10
n = 8 # Número de rectángulos

x = np.linspace(a, b, 100)
x_right = np.linspace(a, b, n+1)
x_right_star = x_right[1:] # Tomamos los extremos derechos

dx = (b - a) / n
y_right = f(x_right_star)
area_right = np.sum(y_right * dx)

plt.plot(x, f(x), 'b-', label='$f(x) = \sin(x) + 1.5$')
plt.bar(x_right_star, y_right, width=-dx, alpha=0.3, align='edge', edgecolor='black', color='red', label='Rectángulos (Derecha)')
plt.title(f'Suma de Riemann (Derecha), Área $\\approx$ {area_right:.3f}')
plt.legend(); plt.show()
```

### Ejemplo 1.2: Aproximación por el Punto Izquierdo

```python
x_left_star = x_right[:-1] # Tomamos los extremos izquierdos
y_left = f(x_left_star)
area_left = np.sum(y_left * dx)

plt.plot(x, f(x), 'b-', label='$f(x)$')
plt.bar(x_left_star, y_left, width=dx, alpha=0.3, align='edge', edgecolor='black', color='green', label='Rectángulos (Izquierda)')
plt.title(f'Suma de Riemann (Izquierda), Área $\\approx$ {area_left:.3f}')
plt.legend(); plt.show()
```

### Ejemplo 1.3: Aproximación por el Punto Medio

```python
x_mid_star = x_left_star + dx/2 # Puntos medios
y_mid = f(x_mid_star)
area_mid = np.sum(y_mid * dx)

plt.plot(x, f(x), 'b-', label='$f(x)$')
plt.bar(x_mid_star, y_mid, width=dx, alpha=0.4, align='center', edgecolor='black', color='orange', label='Rectángulos (Medio)')
plt.plot(x_mid_star, y_mid, 'ro', label='Puntos Medios')
plt.title(f'Suma de Riemann (Punto Medio), Área $\\approx$ {area_mid:.3f}')
plt.legend(); plt.show()
```

# **2. Cálculo de Series Relevantes y el Límite usando SymPy**

Cuando hacemos que el número de rectángulos tienda a infinito ($n \to \infty$), el ancho de cada rectángulo $\Delta x \to 0$. La suma se convierte en la integral definida (área exacta):

$$ \int_{a}^{b} f(x) dx = \lim_{n \to \infty} \sum_{i=1}^{n} f(x_i) \Delta x $$

`SymPy` es excelente para trabajar con sumatorias algebraicas $\sum$ y límites abstractos.

### Ejemplo 2.1: Cálculo de una Serie de Potencias

```python
i, n = sp.symbols('i n', integer=True)

# Serie de la suma de los primeros n cuadrados
serie = sp.Sum(i**2, (i, 1, n))
display(sp.Eq(serie, serie.doit().simplify()))
```

### Ejemplo 2.2: Límite de la Suma de Riemann

```python
# Calculemos el área exacta de f(x) = x^2 en [0, 2] usando la definición
a_val, b_val = 0, 2
delta_x = sp.Rational(b_val - a_val, n)
x_i = a_val + i * delta_x

f_expr = x_i**2
suma_riemann = sp.Sum(f_expr * delta_x, (i, 1, n))

print("Expresión de la Suma de Riemann:")
display(suma_riemann)

# Evaluamos la suma en términos de n
suma_evaluada = suma_riemann.doit()
print("Suma evaluada para un n finito:")
display(suma_evaluada.simplify())

# Tomamos el límite cuando n -> oo
area_exacta = sp.limit(suma_evaluada, n, sp.oo)
print("Área exacta (Límite):")
display(area_exacta)
```

### Ejemplo 2.3: Serie Geométrica en la Ingeniería

```python
# Serie infinita de descarga de un capacitor ideal discretizado
# \sum_{i=0}^{\infty} (1/2)^i
termino = (sp.Rational(1, 2))**i
serie_infinita = sp.Sum(termino, (i, 0, sp.oo))

display(sp.Eq(serie_infinita, serie_infinita.doit()))
```

# **3. Interpretación de la Integral como Antiderivada**

El **Teorema Fundamental del Cálculo (Parte 1)** establece que si $f$ es continua, entonces la función de acumulación de área $A(x) = \int_{a}^{x} f(t) dt$ es continua y diferenciable, tal que:

$$ A'(x) = f(x) $$

Físicamente, si integramos la velocidad, obtenemos el desplazamiento (posición). Si derivamos el desplazamiento, volvemos a la velocidad.

### Ejemplo 3.1: Verificación Simbólica del TFC

```python
t, x_sym = sp.symbols('t x')
f_sym = sp.cos(t**2) # Función cuya integral no es elemental

# Función de acumulación A(x)
A = sp.integrate(f_sym, (t, 0, x_sym))
print("Función Área A(x):")
display(A)

# Derivamos A(x) respecto a x
dA_dx = sp.diff(A, x_sym)
print("Derivada dA/dx:")
display(dA_dx)
# ¡Vuelve a ser la función original f(x) = cos(x^2)!
```

### Ejemplo 3.2: Visualización de la Función de Acumulación

```python
# Veamos gráficamente f(x) y A(x)
x_vals = np.linspace(0, 4*np.pi, 200)
y_f = np.cos(x_vals)

# Acumulamos el área numéricamente (Antiderivada A(x))
y_A = cumtrapz(y_f, x_vals, initial=0)

fig, ax1 = plt.subplots()

ax1.plot(x_vals, y_f, 'b-', label='Función Original $f(x)=\cos(x)$')
ax1.fill_between(x_vals, y_f, color='blue', alpha=0.1)
ax1.set_xlabel('$x$')
ax1.set_ylabel('$f(x)$', color='b')

ax2 = ax1.twinx()
ax2.plot(x_vals, y_A, 'r-', linewidth=3, label='Área Acumulada $A(x)=\sin(x)$')
ax2.set_ylabel('$A(x)$', color='r')

plt.title('Relación Integral-Derivada Gráfica')
fig.legend(loc="upper left", bbox_to_anchor=(0.15,0.85))
plt.show()
```

### Ejemplo 3.3: Aplicación a Cinemática (Velocidad a Posición)

```python
# Simulamos datos ruidosos de la velocidad de un robot en (m/s)
tiempo = np.linspace(0, 10, 50)
vel = 5 * np.sin(tiempo) + np.random.normal(0, 0.5, 50)

# Posición = Integral de la velocidad
posicion = cumtrapz(vel, tiempo, initial=0)

plt.subplot(2, 1, 1)
plt.plot(tiempo, vel, 'g.', label='Velocidad medida (ruidosa)')
plt.ylabel('Velocidad (m/s)')
plt.legend()

plt.subplot(2, 1, 2)
plt.plot(tiempo, posicion, 'purple', linewidth=2, label='Posición integrada (suave)')
plt.ylabel('Posición (m)')
plt.xlabel('Tiempo (s)')
plt.legend()

plt.tight_layout()
plt.show()
```
