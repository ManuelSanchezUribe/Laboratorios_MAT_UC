# **1. El Teorema Fundamental del Cálculo (Analítico)**

El Teorema Fundamental del Cálculo establece el vínculo de oro entre el cálculo diferencial (tasas de cambio) y el cálculo integral (acumulación de área). Se divide en dos partes:

**Parte I (La Derivada de la Acumulación):** Si $f$ es continua en $[a, b]$, entonces la función $g$ definida por la integral de acumulación:
$$g(x) = \int_a^x f(t) dt \quad \text{donde } a \leq x \leq b$$
es continua, derivable, y su derivada es exactamente la función original evaluada en $x$: $g'(x) = f(x)$.

**Parte II (Regla de Evaluación):** Si $f$ es continua, la integral definida se puede calcular mediante cualquier antiderivada $F$ de la función $f$:
$$\int_a^b f(x) dx = F(b) - F(a)$$

Exploraremos estos conceptos valiéndonos del poder simbólico de la librería `sympy`.


### Ejemplo 1.1: Comprobando la Parte I del TFC
Verificaremos que si integramos una función $f(t) = t^2 \sin(t)$ y luego derivamos el resultado respecto al límite superior $x$, obtenemos de regreso la función original. Esto demuestra computacionalmente que $g'(x) = f(x)$.


```python
# Definir símbolos
t, x = sp.symbols('t x')

# Definir la función original (el integrando)
f_t = t**2 * sp.sin(t)

# Calcular la función de acumulación g(x) = integral de f(t) de 0 a x
g_x = sp.integrate(f_t, (t, 0, x))

# Derivar g(x) respecto a x
g_prime = sp.diff(g_x, x)

print("La función de área g(x) es:")
display(g_x)
print("Al derivar g(x) respecto a x, obtenemos:")
display(sp.simplify(g_prime))
```

### Ejemplo 1.2: TFC Parte I con Regla de la Cadena
En problemas más complejos, el límite superior puede ser una función de $x$. Si tenemos $h(x) = \int_a^{u(x)} f(t) dt$, por la regla de la cadena, la derivada será $h'(x) = f(u(x)) \cdot u'(x)$. Evaluemos $\frac{d}{dx} \int_1^{x^3} \cos(t) dt$ usando SymPy.


```python
# Función superior u(x) = x^3
u_x = x**3
f_t2 = sp.cos(t)

# Calcular la integral con límite superior dependiente de x
h_x = sp.integrate(f_t2, (t, 1, u_x))

# Derivamos y simplificamos
h_prime = sp.diff(h_x, x)

print("La derivada de la integral con límite u(x)=x^3 es:")
display(h_prime)
# Nota cómo SymPy aplica matemáticamente la regla de la cadena: cos(x^3) * 3x^2
```

### Ejemplo 1.3: Evaluando Integrales Definidas (Parte II)
Usaremos SymPy para aplicar la Parte II del TFC calculando el área exacta bajo la curva polinomial $f(x) = x^3 - 4x^2 + 3$ en el intervalo $[0, 3]$. Primero hallaremos la antiderivada general $F(x)$, y luego calcularemos $F(3) - F(0)$.


```python
# Definimos la función en términos de x
f_x = x**3 - 4*x**2 + 3

# Paso 1: Obtener la antiderivada (integral indefinida) F(x)
F_x = sp.integrate(f_x, x)
print("Antiderivada F(x):")
display(F_x)

# Paso 2: Evaluar F(3) - F(0)
F_b = F_x.subs(x, 3)
F_a = F_x.subs(x, 0)
area_TFC = F_b - F_a
print(f"Valor evaluando F(3) - F(0) = {area_TFC}")

# Paso 3: Verificación directa con integral definida de SymPy
area_directa = sp.integrate(f_x, (x, 0, 3))
print(f"Valor usando cálculo directo de la integral definida = {area_directa}")
```

# **2. Interpretación Geométrica y Visualización**

La belleza del TFC radica en su significado gráfico. La expresión $\int_a^b f(x) dx$ representa el área neta con signo entre la curva $f(x)$ y el eje $x$. 
Por otro lado, la función $A(x) = \int_a^x f(t) dt$ representa un **área que se acumula** a medida que $x$ avanza de izquierda a derecha. La rapidez con la que se acumula esa área (su derivada) es exactamente la altura de la curva $f(x)$ en ese instante.

Visualicemos esta relación utilizando los métodos numéricos de `numpy` y los gráficos de `matplotlib`.


### Ejemplo 2.1: Sombreado del Área Bajo la Curva
Empezaremos graficando una función gaussiana modificada $f(x) = x^2 e^{-0.5 x}$ y resaltaremos el área que representa la integral definida $\int_1^5 f(x) dx$.


```python
# Crear arreglo numérico de x
x_vals = np.linspace(0, 8, 200)
# Evaluar la función numéricamente
y_vals = x_vals**2 * np.exp(-0.5 * x_vals)

# Configurar el gráfico
plt.figure()
plt.plot(x_vals, y_vals, color='navy', lw=2, label=r'$f(x) = x^2 e^{-0.5 x}$')

# Definir la región de integración [1, 5]
x_fill = np.linspace(1, 5, 100)
y_fill = x_fill**2 * np.exp(-0.5 * x_fill)

# Rellenar el área bajo la curva (Se agregó r antes de la cadena para la etiqueta LaTeX)
plt.fill_between(x_fill, y_fill, color='dodgerblue', alpha=0.4, label=r'Área = $\int_1^5 f(x)dx$')

plt.title("Interpretación Gráfica de la Integral Definida", fontsize=14)
plt.xlabel("x"); plt.ylabel("f(x)")
plt.legend(); plt.show()
```

### Ejemplo 2.2: La Función de Acumulación Interactiva
Utilizaremos `ipywidgets` para crear un gráfico interactivo. Mueve el deslizador superior para modificar el valor de $x$ y observa cómo la función de acumulación $A(x)$ crece a medida que se añade más área bajo la curva $f(t) = \sin(t) + 2$.


```python
def plot_accum(upper_x):
    # Eje t para la función original
    t_vals = np.linspace(0, 10, 200)
    f_vals = np.sin(t_vals) + 2
    
    # Eje para el sombreado [0, upper_x]
    t_fill = np.linspace(0, upper_x, 100)
    f_fill = np.sin(t_fill) + 2
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 6), sharex=True)
    
    # Gráfico superior: Función f(t) y área
    ax1.plot(t_vals, f_vals, 'g-', lw=2, label=r'$f(t) = \sin(t) + 2$')
    ax1.fill_between(t_fill, f_fill, color='lightgreen', alpha=0.5)
    ax1.set_ylabel("f(t)")
    ax1.legend(loc='upper right')
    ax1.set_title(f"Área acumulada hasta x = {upper_x:.2f}")
    
    # Gráfico inferior: Función de acumulación A(x)
    x_hist = np.linspace(0, upper_x, 100)
    # A(x) = -cos(x) + 2x - (-cos(0) + 0) = -cos(x) + 2x + 1
    A_vals = -np.cos(x_hist) + 2*x_hist + 1
    
    # Curva completa de A(x) de fondo tenue
    full_A = -np.cos(t_vals) + 2*t_vals + 1
    ax2.plot(t_vals, full_A, color='gray', linestyle='--', alpha=0.4)
    ax2.plot(x_hist, A_vals, 'r-', lw=3, label=r'$A(x) = \int_0^x f(t)dt$')
    ax2.set_xlabel("x")
    ax2.set_ylabel("A(x)")
    ax2.legend(loc='upper left')
    
    plt.tight_layout()
    plt.show()

# Crear widget interactivo
interact(plot_accum, upper_x=widgets.FloatSlider(min=0.1, max=10.0, step=0.1, value=2.0));
```

### Ejemplo 2.3: Teorema del Valor Medio para Integrales
Directamente relacionado con el TFC, este teorema estipula que existe un punto $c$ en $[a, b]$ tal que el área bajo la curva es igual al área de un rectángulo con base $(b-a)$ y altura $f(c)$ (el valor promedio). Calculemos y grafiquemos el valor promedio de $f(x) = \sqrt{x}$ en el intervalo $[0, 4]$.


```python
# Cálculo analítico del valor promedio
f_sqrt = sp.sqrt(x)
a, b = 0, 4
integral_val = sp.integrate(f_sqrt, (x, a, b))
f_promedio = integral_val / (b - a)

print(f"El valor promedio de f(x) en [{a}, {b}] es: {f_promedio} = {float(f_promedio):.3f}")

# Visualización
x_num = np.linspace(a, b, 100)
y_num = np.sqrt(x_num)

plt.figure()
plt.plot(x_num, y_num, color='darkorange', lw=2, label=r'$f(x) = \sqrt{x}$')
# Graficar el rectángulo de valor promedio
plt.hlines(float(f_promedio), a, b, colors='red', linestyles='dashed', lw=2, label=r'Altura Promedio $f_{avg}$')
plt.fill_between(x_num, float(f_promedio), color='red', alpha=0.1, label='Área del Rectángulo')

plt.title("Teorema del Valor Medio para Integrales")
plt.legend(loc='upper left')
plt.show()
```

# **3. Aplicaciones en Ingeniería: El Teorema del Cambio Neto**

El TFC proporciona la herramienta fundamental para el modelaje en ingeniería. Si conocemos la tasa de cambio de cualquier cantidad $Q(t)$ (es decir, conocemos su derivada $Q'(t)$), la integral de esa tasa sobre un intervalo de tiempo nos da el **cambio neto** en la cantidad:

$$\int_{t_1}^{t_2} Q'(t) dt = Q(t_2) - Q(t_1)$$

* En **Cinemática**: $\int v(t)dt$ da el desplazamiento y $\int |v(t)|dt$ la distancia.
* En **Hidráulica**: $\int Q_v(t) dt$ da el volumen acumulado de flujo.
* En **Dinámica**: $\int F(x)dx$ da el trabajo mecánico realizado.


### Ejemplo 3.1: Cinemática (Desplazamiento vs Distancia Total)
Una partícula se mueve a lo largo de una línea con velocidad $v(t) = t^2 - t - 6$ (m/s). Analizaremos el movimiento entre $t = 1$ y $t = 4$. 
Recordemos que el *desplazamiento* puede cancelar áreas positivas y negativas, mientras que la *distancia total* requiere el valor absoluto $|v(t)|$.


```python
v_t = t**2 - t - 6

# Desplazamiento (integral simple)
desplazamiento = sp.integrate(v_t, (t, 1, 4))

# Para la distancia total, debemos integrar el valor absoluto.
# Buscamos donde v(t) = 0 para separar las integrales (Raíces de la parábola)
raices = sp.solve(v_t, t)
print(f"La velocidad es cero en t = {raices}")
# Raíces en t=-2 y t=3. En [1, 4], el signo cambia en t=3.

# Distancia total = Integral de 1 a 3 de -v(t) + Integral de 3 a 4 de v(t)
dist_1_3 = sp.integrate(-v_t, (t, 1, 3))
dist_3_4 = sp.integrate(v_t, (t, 3, 4))
distancia_total = dist_1_3 + dist_3_4

print(f"El desplazamiento neto es: {desplazamiento} metros")
print(f"La distancia total recorrida es: {distancia_total} metros")
```

### Ejemplo 3.2: Acumulación de Flujo en un Tanque (Ingeniería Química)
Un tanque industrial contiene un producto químico líquido. Debido a una filtración, el químico se derrama a una tasa de $R(t) = 150 e^{-0.1t}$ litros/hora. ¿Cuál es el volumen total de líquido que se ha derramado durante las primeras 10 horas?


```python
# Definir la función de tasa de derrame
R_t = 150 * sp.exp(-0.1 * t)

# El volumen derramado es la integral de la tasa de flujo (Cambio Neto)
Volumen_derramado = sp.integrate(R_t, (t, 0, 10))

print("Volumen derramado de forma analítica exacta:")
display(Volumen_derramado)

# Aproximación numérica
volumen_num = Volumen_derramado.evalf()
print(f"Aproximadamente: {volumen_num:.2f} Litros perdidos en 10 horas.")
```

### Ejemplo 3.3: Trabajo Mecánico de una Fuerza Variable
En física e ingeniería mecánica, el Trabajo $W$ realizado por una fuerza que mueve un objeto a lo largo del eje x desde $a$ hasta $b$ es $W = \int_a^b F(x) dx$. 
Calcularemos el trabajo necesario para comprimir un resorte industrial no lineal, cuya fuerza de resistencia está dada por $F(x) = 400x + 150x^3$ (Newtons), desde su posición de equilibrio $x=0$ hasta $x=0.5$ metros.


```python
F_x = 400*x + 150*x**3

# Cálculo del trabajo mecánico
Trabajo = sp.integrate(F_x, (x, 0, 0.5))

print(f"El trabajo mecánico (energía requerida) para comprimir el resorte es: {Trabajo} Joules")
```
