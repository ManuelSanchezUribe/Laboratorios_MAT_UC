# **1. Problemas Aplicados de Optimización**

En ingeniería, rara vez nos dan una función explícita $f(x)$ para derivar. A menudo enfrentamos un problema del mundo real (ej. "minimizar el material de un tanque" o "maximizar la resistencia de una viga"). La modelación consiste en **construir esa función objetivo** utilizando la geometría y luego reducirla a una sola variable utilizando ecuaciones de **restricción**.

Recordemos que, analíticamente, el valor máximo o mínimo de una función continua en un intervalo cerrado $[a, b]$ ocurre ya sea en un **punto crítico** (donde $f'(x) = 0$ o $f'(x)$ no existe) o en los **extremos** del intervalo (Fermat).

## 1.1 Intuición Geométrica y Traducción al Código
Nuestro flujo de trabajo en Python será:
1. Definir variables simbólicas con `SymPy`.
2. Establecer la ecuación de restricción.
3. Crear la función objetivo de una sola variable.
4. Calcular la derivada y encontrar sus raíces.
5. Graficar para confirmar visualmente el resultado.


### Ejemplo 1.1: Maximización de Volumen (Diseño de Empaques)
Una fábrica necesita construir una caja rectangular sin tapa a partir de una lámina de cartón de $30 \text{ cm}$ por $20 \text{ cm}$. Para ello, cortará cuadrados idénticos de lado $x$ en cada esquina y doblará las solapas hacia arriba. ¿Cuál debe ser el valor de $x$ para que la caja contenga el **volumen máximo**?


```python
# 1. Definimos la variable simbólica
x = sp.Symbol('x', real=True, positive=True)

# 2. Modelamos las dimensiones de la caja
L = 30 - 2*x  # Largo
W = 20 - 2*x  # Ancho
H = x         # Altura

# 3. Función Objetivo: Volumen
V = L * W * H
display(sp.Eq(sp.Symbol('V(x)'), V))

# 4. Derivada y Puntos Críticos
dV_dx = sp.diff(V, x)
puntos_criticos = sp.solve(dV_dx, x)

print("Posibles valores para x (puntos críticos):", puntos_criticos)
# Notamos que x no puede ser mayor a 10 (ya que W = 20 - 2x debe ser positivo)
x_optimo = float(puntos_criticos[0]) # Seleccionamos la raíz lógica
volumen_max = float(V.subs(x, x_optimo))
print(f"El corte óptimo es x ≈ {x_optimo:.2f} cm, generando un volumen máximo de {volumen_max:.2f} cm³")

# 5. Visualización para confirmación
x_vals = np.linspace(0, 10, 100) # Dominio con sentido físico
V_lambdify = sp.lambdify(x, V, 'numpy')

plt.plot(x_vals, V_lambdify(x_vals), color='#2980b9', lw=2)
plt.plot(x_optimo, volumen_max, 'ro', label=f'Máximo ({x_optimo:.2f}, {volumen_max:.2f})')
plt.title("Volumen de la caja en función del corte x")
plt.xlabel("$x$ (cm)")
plt.ylabel("Volumen $V(x)$ ($cm^3$)")
plt.legend()
plt.show()
```

### Ejemplo 1.2: Minimización de Costos Industriales (Tanque Cilíndrico)
Se desea diseñar un tanque de almacenamiento cilíndrico cerrado capaz de contener $2000 \text{ m}^3$ de líquido químico. El material para la base y la tapa cuesta $\$40$ por metro cuadrado, mientras que el material de la cara lateral cuesta $\$25$ por metro cuadrado. Determine el radio $r$ y la altura $h$ que **minimizan el costo** de construcción.


```python
# 1. Variables
r, h = sp.symbols('r h', real=True, positive=True)

# 2. Restricción (Volumen fijo = 2000)
# V = pi * r^2 * h = 2000  =>  h = 2000 / (pi * r^2)
h_expr = 2000 / (sp.pi * r**2)

# 3. Función Objetivo (Costo)
# C = Area_tapas * Costo_tapas + Area_lateral * Costo_lateral
costo_tapas = (2 * sp.pi * r**2) * 40
costo_lateral = (2 * sp.pi * r * h) * 25

C_rh = costo_tapas + costo_lateral
# Sustituimos h para tener solo una variable r
C = C_rh.subs(h, h_expr)
display(sp.Eq(sp.Symbol('C(r)'), sp.simplify(C)))

# 4. Derivar e igualar a 0
dC_dr = sp.diff(C, r)
r_optimo_simb = sp.solve(dC_dr, r)[0] # Sympy halla la raíz exacta
r_optimo = float(r_optimo_simb.evalf())
h_optimo = float(h_expr.subs(r, r_optimo).evalf())
costo_min = float(C.subs(r, r_optimo).evalf())

print(f"Dimensiones óptimas: Radio = {r_optimo:.2f} m, Altura = {h_optimo:.2f} m")
print(f"Costo Mínimo = ${costo_min:,.2f}")
```

### Ejemplo 1.3: Minimización de Distancias (Trayectoria de Drones)
Un dron se desplaza a lo largo de una trayectoria parabólica modelada por la función $y = 4 - x^2$. Una torre de control está ubicada en el punto $(0, 6)$. ¿Cuál es la distancia mínima entre el dron y la torre? 

*Nota Analítica:* Minimizar la distancia $D = \sqrt{\Delta x^2 + \Delta y^2}$ es matemáticamente equivalente a minimizar la distancia al cuadrado $D^2$. Esto nos ahorra derivar raíces cuadradas.


```python
# Variables
x = sp.Symbol('x', real=True)
y_curva = 4 - x**2
x_torre, y_torre = 0, 6

# Función objetivo: Distancia al cuadrado
D_cuadrado = (x - x_torre)**2 + (y_curva - y_torre)**2
display(sp.Eq(sp.Symbol('D^2(x)'), sp.expand(D_cuadrado)))

# Encontrar puntos críticos
dD2_dx = sp.diff(D_cuadrado, x)
raices = sp.solve(dD2_dx, x)

print("Raíces de la derivada:", raices)
# Vemos que x = 0 o x = +/- sqrt(3/2)
# Evaluamos las distancias reales (con la raíz)
distancias = [float(sp.sqrt(D_cuadrado).subs(x, val)) for val in raices]
min_dist = min(distancias)

print(f"La distancia mínima es {min_dist:.2f} unidades.")

# Gráfico para validar la intuición
x_vals = np.linspace(-3, 3, 200)
y_vals = 4 - x_vals**2

plt.plot(x_vals, y_vals, label="Trayectoria del Dron ($y = 4 - x^2$)")
plt.plot(0, 6, 'k^', markersize=10, label="Torre de Control (0,6)")

# Trazar línea de distancia mínima (usamos la raíz positiva para visualizar una)
x_opt = float(raices[1]) # sqrt(3/2)
plt.plot([0, x_opt], [6, 4 - x_opt**2], 'r--', label="Distancia mínima")

plt.xlim(-4, 4)
plt.ylim(0, 8)
plt.gca().set_aspect('equal')
plt.legend()
plt.show()
```

# **2. Aplicaciones de la Integral en Ingeniería**

La integral definida $$ \int_{a}^{b} f(x) dx $$ se interpreta geométricamente como el área bajo la curva. Sin embargo, en un sentido más amplio, la integral es un **acumulador**. Nos permite calcular volúmenes de sólidos tridimensionales (método de discos/arandelas) o áreas de superficies curvas.

Las dos fórmulas de modelación más importantes en la fabricación y diseño de piezas que giran sobre un eje (tornillos, pistones, antenas) son:

1. **Volumen de Revolución (Método de Discos):**
$$ V = \int_{a}^{b} \pi [f(x)]^2 dx $$

2. **Área de una Superficie de Revolución:**
$$ S = \int_{a}^{b} 2\pi f(x) \sqrt{1 + [f'(x)]^2} dx $$


### Ejemplo 2.1: Cálculo de Volumen de Revolución (Método de Discos)
Un ingeniero está diseñando una válvula mecánica cuya forma se obtiene al rotar la región acotada por $y = \sqrt{x}$, el eje $x$, y las rectas $x=0$ y $x=4$, alrededor del eje $x$. Calcule el volumen exacto de esta pieza utilizando integración simbólica.


```python
x = sp.Symbol('x')
f = sp.sqrt(x)

# Formula del volumen: V = pi * integral(f(x)^2) dx
integrando_V = sp.pi * f**2
Volumen = sp.integrate(integrando_V, (x, 0, 4))

display(sp.Eq(sp.Symbol('Volumen'), Volumen))
print(f"El volumen numérico es: {float(Volumen.evalf()):.2f} unidades cúbicas")

# --- Visualización 3D ---
u = np.linspace(0, 4, 100)
v = np.linspace(0, 2*np.pi, 100)
U, V = np.meshgrid(u, v)

# Parametrización para rotar sobre el eje X
X = U
Y = np.sqrt(U) * np.cos(V)
Z = np.sqrt(U) * np.sin(V)

fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, cmap='plasma', alpha=0.8)
ax.set_title("Válvula Generada por Revolución")
ax.set_xlabel('Eje X')
plt.show()
```

### Ejemplo 2.2: Área de Superficie Curva (Antena Parabólica)
Se desea recubrir con una pintura especial el plato de una antena parabólica. La antena se describe rotando la curva $y = \frac{x^3}{3}$ alrededor del eje $x$, desde $x=0$ hasta $x=2$. 
Determine el **área de la superficie** a pintar.

*Consejo Pedagógico:* Muchas integrales de superficie resultan muy complejas para resolver a mano. Utilizaremos las capacidades de cálculo numérico integradas en `SciPy` (`scipy.integrate.quad`) para estas modelaciones cuando `SymPy` no encuentra soluciones analíticas rápidas.


```python
# 1. Enfoque Simbólico
x = sp.Symbol('x')
y_func = (x**3) / 3
dy_dx = sp.diff(y_func, x)

# S = integral( 2*pi * f(x) * sqrt(1 + f'(x)^2) )
integrando_S = 2 * sp.pi * y_func * sp.sqrt(1 + dy_dx**2)
display(sp.Eq(sp.Symbol('Integral_Superficie'), sp.Integral(integrando_S, (x, 0, 2))))

# Evaluamos numéricamente usando integración de SciPy para mayor rapidez
# Convertimos la función simbólica a una lambda de numpy
integrando_num = sp.lambdify(x, integrando_S, modules=['numpy'])

# scipy.integrate.quad retorna el valor de la integral y el error estimado
Area_Superficie, error = spi.quad(integrando_num, 0, 2)

print(f"El área de superficie que se debe pintar es aproximadamente: {Area_Superficie:.4f} unidades cuadradas.")
```

### Ejemplo 2.3: Longitud de Arco (Diseño de Puentes Colgantes)
En un puente colgante, la forma que adquiere el cable principal bajo su propio peso es descrita por una **catenaria**, modelada matemáticamente por $y = a \cosh\left(\frac{x}{a}\right)$. 
Si tenemos un cable con parámetro $a=50$ que cuelga simétricamente entre dos torres situadas en $x=-50$ y $x=50$, calcule la longitud exacta del cable usando la fórmula de longitud de arco:
$$ L = \int_{a}^{b} \sqrt{1 + [f'(x)]^2} dx $$


```python
# Parámetros del diseño
a = 50
x = sp.Symbol('x')
f_catenaria = a * sp.cosh(x / a)

# Derivada
df_catenaria = sp.diff(f_catenaria, x)

# Simplificación del integrando: 1 + sinh^2(x/a) = cosh^2(x/a)
# Sympy puede simplificar esto algebraicamente
radicando = 1 + df_catenaria**2
integrando_L = sp.sqrt(radicando).simplify()

display(sp.Eq(sp.Symbol('Integrando_Simplificado'), integrando_L))

# Calculamos la integral analítica
Longitud_cable = sp.integrate(integrando_L, (x, -50, 50))
Longitud_cable_val = float(Longitud_cable.evalf())

print(f"La longitud exacta del cable es de {Longitud_cable_val:.2f} metros.")

# Trazado de la catenaria
x_plot = np.linspace(-50, 50, 100)
y_plot = a * np.cosh(x_plot / a)

plt.plot(x_plot, y_plot, 'k-', lw=3, label="Cable Catenaria")
plt.plot([-50, -50], [0, a*np.cosh(-1)], 'gray', lw=5, label="Torres")
plt.plot([50, 50], [0, a*np.cosh(1)], 'gray', lw=5)
plt.fill_between(x_plot, 0, y_plot, color='skyblue', alpha=0.2)
plt.title("Perfil del Cable Colgante")
plt.legend()
plt.show()
```
